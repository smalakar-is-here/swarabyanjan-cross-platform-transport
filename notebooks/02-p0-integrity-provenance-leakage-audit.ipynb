## Package discovery, cryptographic verification, and finding utilities

````python

SEVERITY_ORDER = {
    "BLOCKER": 0,
    "MAJOR": 1,
    "MINOR": 2,
    "INFORMATIONAL": 3,
}
FINDINGS = []
EXECUTION_LOG = []

def utc_now():
    return datetime.now(timezone.utc).isoformat()

def log(message):
    line = f"{utc_now()} | {message}"
    EXECUTION_LOG.append(line)
    print(line)

def add_finding(severity, code, domain, message, stage03_blocking=False, evidence=None):
    if severity not in SEVERITY_ORDER:
        raise ValueError(severity)
    FINDINGS.append({
        "severity": severity,
        "code": code,
        "domain": domain,
        "stage03_blocking": bool(stage03_blocking),
        "message": message,
        "evidence": evidence or {},
    })

def sha256_file(path, chunk_size=8 * 1024 * 1024):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                return h.hexdigest()
            h.update(chunk)

def normalize_tree_relpath(rel):
    raw = str(rel)
    if not raw or raw != raw.strip():
        raise RuntimeError(f"Invalid tree-relative path: {rel!r}")
    if "\\" in raw:
        raise RuntimeError(f"Backslash is prohibited in frozen tree paths: {raw!r}")
    posix = PurePosixPath(raw)
    if posix.is_absolute():
        raise RuntimeError(f"Absolute path prohibited in frozen tree: {raw!r}")
    if any(part in {"", ".", ".."} for part in posix.parts):
        raise RuntimeError(f"Non-canonical tree-relative path: {raw!r}")
    normalized = posix.as_posix()
    if normalized != raw:
        raise RuntimeError(
            f"Tree path is not canonical POSIX form: {raw!r} -> {normalized!r}"
        )
    return normalized


def validate_full_sha256(value, field_name):
    value = str(value).strip().lower()
    if not re.fullmatch(r"[0-9a-f]{64}", value):
        raise RuntimeError(
            f"{field_name} must be a full 64-character lowercase/uppercase "
            f"hexadecimal SHA-256 value; received {value!r}"
        )
    return value


def canonical_tree_fingerprint_from_sums(sums):
    canonical_lines = [
        f"{sums[rel]}  {rel}"
        for rel in sorted(sums)
    ]
    payload = ("\n".join(canonical_lines) + "\n").encode("utf-8")
    return hashlib.sha256(payload).hexdigest()


def verify_exact_tree_member_set(root, sums):
    root = Path(root).resolve()
    actual_files = []
    for path in root.rglob("*"):
        if path.is_symlink():
            raise RuntimeError(f"Symlink prohibited inside frozen tree: {path}")
        if path.is_file():
            rel = normalize_tree_relpath(path.relative_to(root).as_posix())
            actual_files.append(rel)

    actual_set = set(actual_files)
    expected_set = set(sums) | {"SHA256SUMS.txt"}

    unexpected = sorted(actual_set - expected_set)
    missing = sorted(expected_set - actual_set)
    if unexpected or missing:
        raise RuntimeError(
            "Frozen tree member-set mismatch; "
            f"unexpected={unexpected}; missing={missing}"
        )
    return sorted(actual_set)








def kaggle_mount_root(path, input_root):
    input_root = Path(input_root).resolve()
    path = Path(path).resolve()

    try:
        rel = path.relative_to(input_root)
    except ValueError as exc:
        raise RuntimeError(
            f"Path is outside canonical Kaggle input root: {path}"
        ) from exc

    if not rel.parts:
        raise RuntimeError(
            f"Artifact is not under a mounted Kaggle Dataset: {path}"
        )

    return (input_root / rel.parts[0]).resolve()




def parse_sha256_sidecar(path, expected_filename):
    text = Path(path).read_text(
        encoding="utf-8",
        errors="strict",
    ).strip()
    parts = text.split()
    if not parts:
        raise RuntimeError(f"Empty SHA-256 sidecar: {path}")

    digest = validate_full_sha256(
        parts[0],
        f"sidecar digest in {path}",
    )
    if len(parts) > 1:
        stated_name = Path(parts[-1].lstrip("*")).name
        if stated_name != expected_filename:
            raise RuntimeError(
                "SHA-256 sidecar names a different file: "
                f"{stated_name!r} != {expected_filename!r}"
            )

    return digest


def discover_canonical_zip_mount(
    input_root,
    required=False,
):
    input_root = Path(input_root).resolve()

    zip_candidates = sorted(
        input_root.rglob(FREEZE_ID + ".zip"),
        key=lambda p: str(p),
    )
    if not zip_candidates:
        if required:
            raise RuntimeError(
                f"Missing canonical {FREEZE_ID}.zip under {input_root}"
            )
        return None

    valid = []
    invalid = []

    for zip_path in zip_candidates:
        digest = sha256_file(zip_path)
        if digest != EXPECTED_ZIP_SHA256:
            invalid.append((zip_path, digest))
            continue

        mount_root = kaggle_mount_root(zip_path, input_root)
        sidecars = sorted(
            mount_root.rglob(zip_path.name + ".sha256"),
            key=lambda p: str(p),
        )
        matching_sidecars = []
        invalid_sidecars = []

        for sidecar in sidecars:
            try:
                sidecar_digest = parse_sha256_sidecar(
                    sidecar,
                    zip_path.name,
                )
                if sidecar_digest == EXPECTED_ZIP_SHA256:
                    matching_sidecars.append(sidecar)
                else:
                    invalid_sidecars.append(
                        (sidecar, sidecar_digest)
                    )
            except Exception as exc:
                invalid_sidecars.append(
                    (
                        sidecar,
                        f"{type(exc).__name__}: {exc}",
                    )
                )

        if invalid_sidecars:
            raise RuntimeError(
                "Invalid canonical ZIP sidecar candidate(s) within "
                f"mount {mount_root}: "
                + " | ".join(
                    f"{p}={d}"
                    for p, d in invalid_sidecars
                )
            )

        if len(matching_sidecars) != 1:
            raise RuntimeError(
                "Exactly one matching canonical ZIP sidecar is required "
                f"within mount {mount_root}; "
                f"matching_count={len(matching_sidecars)}"
            )

        valid.append({
            "representation": "ZIP",
            "zip_path": zip_path.resolve(),
            "zip_sha256": digest,
            "zip_sidecar_path": matching_sidecars[0].resolve(),
            "mount_root": mount_root,
        })

    if invalid:
        raise RuntimeError(
            f"Non-canonical file(s) named {FREEZE_ID}.zip were supplied: "
            + " | ".join(
                f"{p}={d}"
                for p, d in invalid
            )
        )

    if len(valid) != 1:
        raise RuntimeError(
            "Exactly one canonical literal ZIP is permitted across "
            f"{input_root}; matching_count={len(valid)}"
        )

    return valid[0]

def discover_canonical_extracted_root(
    input_root,
    required=False,
):
    input_root = Path(input_root).resolve()

    candidates = sorted(
        [
            p.resolve()
            for p in input_root.rglob(FREEZE_ID)
            if p.is_dir()
        ],
        key=lambda p: str(p),
    )

    if not candidates:
        if required:
            raise RuntimeError(
                f"Missing canonical extracted root {FREEZE_ID}/ "
                f"under {input_root}"
            )
        return None

    candidate_evaluations = []
    valid = []

    for root in candidates:
        evaluation = {
            "candidate_path": str(root),
            "candidate_basename": root.name,
            "status": "REJECTED",
            "rejection_reason": None,
            "structural_check": None,
            "internal_integrity_check": None,
            "source_identity_check": None,
            "sidecar_check": None,
            "canonical_zip_anchor_check": None,
        }

        try:
            mount_root = kaggle_mount_root(
                root,
                input_root,
            )
            evaluation["mount_root"] = str(mount_root)

            # First distinguish wrapper directories from actual package roots.
            # Matching the basename alone is never sufficient.
            missing_controls = sorted(
                name
                for name in CONTROL_FILES
                if not (root / name).is_file()
            )
            missing_dataset_dirs = sorted(
                name
                for name in REQUIRED_DATASET_DIRS
                if not (root / name).is_dir()
            )

            if missing_controls or missing_dataset_dirs:
                evaluation["structural_check"] = {
                    "status": "FAIL",
                    "missing_control_files": missing_controls,
                    "missing_dataset_directories":
                        missing_dataset_dirs,
                }
                raise RuntimeError(
                    "Structural package-root check failed: "
                    f"missing_control_files={missing_controls}; "
                    f"missing_dataset_directories={missing_dataset_dirs}"
                )

            evaluation["structural_check"] = {
                "status": "PASS",
                "missing_control_files": [],
                "missing_dataset_directories": [],
            }

            # This verifies exact tree membership, every SHA256SUMS entry,
            # every manifest-declared required file, dataset status records,
            # and the complete manifest/file consistency contract.
            manifest_record = verify_manifest_root(root)
            evaluation["internal_integrity_check"] = {
                "status": "PASS",
                "content_tree_fingerprint":
                    manifest_record[
                        "content_tree_fingerprint"
                    ],
                "verified_tree_member_count":
                    len(
                        manifest_record[
                            "verified_tree_members"
                        ]
                    ),
                "verified_file_count":
                    len(manifest_record["verified_files"]),
            }

            # Immutable source identity is part of candidate validity, not a
            # post-selection convenience check.
            source_identity_checks = verify_source_identities(
                manifest_record
            )
            evaluation["source_identity_check"] = {
                "status": "PASS",
                "checks": source_identity_checks,
            }

            sidecars = sorted(
                mount_root.rglob(
                    FREEZE_ID + ".zip.sha256"
                ),
                key=lambda p: str(p),
            )
            matching_sidecars = []
            invalid_sidecars = []

            for sidecar in sidecars:
                try:
                    sidecar_digest = parse_sha256_sidecar(
                        sidecar,
                        FREEZE_ID + ".zip",
                    )
                    if sidecar_digest == EXPECTED_ZIP_SHA256:
                        matching_sidecars.append(sidecar)
                    else:
                        invalid_sidecars.append({
                            "path": str(sidecar),
                            "reason": (
                                "digest_mismatch:"
                                + str(sidecar_digest)
                            ),
                        })
                except Exception as exc:
                    invalid_sidecars.append({
                        "path": str(sidecar),
                        "reason":
                            f"{type(exc).__name__}: {exc}",
                    })

            if invalid_sidecars:
                evaluation["sidecar_check"] = {
                    "status": "FAIL",
                    "matching_sidecars": [
                        str(p)
                        for p in matching_sidecars
                    ],
                    "invalid_sidecars":
                        sorted(
                            invalid_sidecars,
                            key=lambda x: x["path"],
                        ),
                }
                raise RuntimeError(
                    "Invalid frozen ZIP sidecar candidate(s) within "
                    f"mount {mount_root}: "
                    + " | ".join(
                        f"{item['path']}={item['reason']}"
                        for item in sorted(
                            invalid_sidecars,
                            key=lambda x: x["path"],
                        )
                    )
                )

            if len(matching_sidecars) != 1:
                evaluation["sidecar_check"] = {
                    "status": "FAIL",
                    "matching_sidecars": [
                        str(p)
                        for p in matching_sidecars
                    ],
                    "invalid_sidecars": [],
                }
                raise RuntimeError(
                    "Exactly one matching frozen ZIP SHA-256 sidecar "
                    "is required within the extracted-root mount; "
                    f"matching_count={len(matching_sidecars)}"
                )

            evaluation["sidecar_check"] = {
                "status": "PASS",
                "matching_sidecar_path":
                    str(matching_sidecars[0].resolve()),
                "expected_zip_sha256":
                    EXPECTED_ZIP_SHA256,
            }

            # External trust anchor. Internal controls alone are not enough:
            # the complete candidate root must reconstruct byte-for-byte to the
            # canonical Stage-01 ZIP identity. The temporary ZIP is deleted by
            # verify_extracted_root_against_canonical_zip().
            anchor = verify_extracted_root_against_canonical_zip(
                root,
                EXPECTED_ZIP_SHA256,
                TEMP_ROOT,
            )
            evaluation["canonical_zip_anchor_check"] = {
                "status": anchor["status"],
                "method": anchor["method"],
                "expected_zip_sha256":
                    anchor["expected_zip_sha256"],
                "verified_reconstructed_zip_sha256":
                    anchor[
                        "verified_reconstructed_zip_sha256"
                    ],
                "member_count": anchor["member_count"],
            }

            evaluation["status"] = "VERIFIED_CANONICAL"
            evaluation["rejection_reason"] = None

            valid.append({
                "representation": "EXTRACTED_ROOT",
                "root": root,
                "mount_root": mount_root,
                "zip_sidecar_path":
                    matching_sidecars[0].resolve(),
                "manifest_record": manifest_record,
                "source_identity_checks":
                    source_identity_checks,
                "tree_anchor": anchor,
                "candidate_evaluation": evaluation,
            })

        except Exception as exc:
            evaluation["status"] = "REJECTED"
            evaluation["rejection_reason"] = (
                f"{type(exc).__name__}: {exc}"
            )

        candidate_evaluations.append(evaluation)

    # Candidate rejection is diagnostic evidence, not an automatic global
    # failure. A nested valid root must remain eligible even when a same-named
    # outer Kaggle/upload wrapper is structurally invalid.
    if len(valid) == 0:
        raise RuntimeError(
            "No valid canonical extracted frozen root was found after "
            f"evaluating {len(candidates)} candidate(s). "
            "Candidate results: "
            + json.dumps(
                sorted(
                    candidate_evaluations,
                    key=lambda x: x["candidate_path"],
                ),
                ensure_ascii=False,
                sort_keys=True,
                separators=(",", ":"),
            )
        )

    if len(valid) > 1:
        raise RuntimeError(
            "Ambiguous extracted frozen input: more than one candidate "
            "passed the complete canonical package and ZIP-anchor checks. "
            "Verified roots: "
            + json.dumps(
                sorted(
                    [
                        str(item["root"])
                        for item in valid
                    ]
                ),
                ensure_ascii=False,
                separators=(",", ":"),
            )
            + "; candidate results: "
            + json.dumps(
                sorted(
                    candidate_evaluations,
                    key=lambda x: x["candidate_path"],
                ),
                ensure_ascii=False,
                sort_keys=True,
                separators=(",", ":"),
            )
        )

    accepted = valid[0]
    accepted_path = str(
        Path(accepted["root"]).resolve()
    )

    accepted["candidate_paths_examined"] = [
        item["candidate_path"]
        for item in sorted(
            candidate_evaluations,
            key=lambda x: x["candidate_path"],
        )
    ]
    accepted["candidate_evaluations"] = sorted(
        candidate_evaluations,
        key=lambda x: x["candidate_path"],
    )
    accepted["accepted_canonical_root_path"] = (
        accepted_path
    )
    accepted["canonical_verification_result"] = {
        "status": "PASS",
        "internal_integrity":
            accepted["candidate_evaluation"][
                "internal_integrity_check"
            ],
        "source_identity":
            accepted["candidate_evaluation"][
                "source_identity_check"
            ],
        "zip_sidecar":
            accepted["candidate_evaluation"][
                "sidecar_check"
            ],
        "canonical_zip_anchor":
            accepted["candidate_evaluation"][
                "canonical_zip_anchor_check"
            ],
    }

    return accepted


def materialize_and_verify_canonical_zip(
    zip_record,
):
    zp = Path(zip_record["zip_path"])
    sidecar = Path(zip_record["zip_sidecar_path"])

    if TEMP_ROOT.exists():
        shutil.rmtree(TEMP_ROOT)
    extract_parent = TEMP_ROOT / "verified_zip_extraction"
    safe_extract_zip(zp, extract_parent)

    root_candidates = [
        p.parent
        for p in extract_parent.rglob("SOURCES_LOCK.json")
        if (p.parent / "ACQUISITION_MANIFEST.json").is_file()
    ]
    root_candidates = sorted(
        set(root_candidates),
        key=lambda p: str(p),
    )
    if len(root_candidates) != 1:
        raise RuntimeError(
            "Canonical ZIP must contain exactly one frozen package root; "
            f"candidate_count={len(root_candidates)}"
        )

    root = root_candidates[0]
    rec = verify_manifest_root(root)

    return {
        "root": root,
        "manifest_record": rec,
        "zip_path": zp,
        "zip_sidecar_path": sidecar,
        "mount_root": Path(zip_record["mount_root"]),
    }



def enforce_stage03_field_firewall(
    requested_text_fields,
    requested_target_fields,
    firewall,
    expected_firewall_sha256,
):
    expected_firewall_sha256 = validate_full_sha256(
        expected_firewall_sha256,
        "expected_firewall_sha256",
    )
    actual_firewall_sha256 = hashlib.sha256(
        json.dumps(
            firewall,
            sort_keys=True,
            ensure_ascii=False,
            separators=(",", ":"),
        ).encode("utf-8")
    ).hexdigest()
    if actual_firewall_sha256 != expected_firewall_sha256:
        raise RuntimeError(
            "Stage-03 firewall SHA-256 mismatch; a versioned firewall revision "
            "is required before use."
        )

    contract = firewall.get("stage03_model_input_contract", {})
    if contract.get("strict") is not True:
        raise RuntimeError("Stage-03 model-input firewall is not strict")

    requested_text = sorted({str(x) for x in requested_text_fields})
    requested_targets = sorted({str(x) for x in requested_target_fields})
    allowed_text = sorted({str(x) for x in contract.get("allowed_text_fields", [])})
    allowed_targets = sorted({str(x) for x in contract.get("allowed_target_fields", [])})

    disallowed_text = sorted(set(requested_text) - set(allowed_text))
    disallowed_targets = sorted(set(requested_targets) - set(allowed_targets))
    if disallowed_text or disallowed_targets:
        raise RuntimeError(
            "Stage-03 requested fields are not authorized by the frozen "
            f"Stage-02 firewall; text={disallowed_text}; "
            f"targets={disallowed_targets}"
        )

    return {
        "firewall_sha256": actual_firewall_sha256,
        "requested_text_fields": requested_text,
        "requested_target_fields": requested_targets,
        "allowed_text_fields": allowed_text,
        "allowed_target_fields": allowed_targets,
        "pass": True,
    }

def parse_sha256sums(path):
    result = {}
    for line in Path(path).read_text(encoding="utf-8").splitlines():
        if not line.strip():
            continue
        digest, rel = line.split(None, 1)
        rel = rel.lstrip("*").strip()
        if not re.fullmatch(r"[0-9a-fA-F]{64}", digest):
            raise RuntimeError(f"Malformed SHA256SUMS digest: {line}")
        rel = normalize_tree_relpath(rel)
        if rel in result:
            raise RuntimeError(f"Duplicate SHA256SUMS path: {rel}")
        result[rel] = digest.lower()
    if not result:
        raise RuntimeError("SHA256SUMS.txt contains no entries")
    return result

def safe_child(root, rel):
    root = Path(root).resolve()
    target = (root / rel).resolve()
    if target != root and root not in target.parents:
        raise RuntimeError(f"Unsafe relative path: {rel}")
    return target

def safe_extract_zip(zip_path, destination):
    destination = Path(destination).resolve()
    destination.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path, "r") as zf:
        for info in zf.infolist():
            target = (destination / info.filename).resolve()
            if target != destination and destination not in target.parents:
                raise RuntimeError(f"Unsafe ZIP member: {info.filename}")
        zf.extractall(destination)
        bad = zf.testzip()
        if bad is not None:
            raise RuntimeError(f"ZIP CRC failure: {bad}")

def read_json(path):
    return json.loads(Path(path).read_text(encoding="utf-8"))

def sidecar_matches(zip_path):
    zip_path = Path(zip_path)
    candidates = [zip_path.with_name(zip_path.name + ".sha256")]
    candidates += sorted(INPUT_ROOT.rglob(zip_path.name + ".sha256"))
    seen = set()
    for candidate in candidates:
        key = str(candidate.resolve()) if candidate.exists() else str(candidate)
        if key in seen or not candidate.is_file():
            continue
        seen.add(key)
        text = candidate.read_text(encoding="utf-8", errors="strict").strip()
        parts = text.split()
        if not parts:
            continue
        if parts[0].lower() == EXPECTED_ZIP_SHA256:
            if len(parts) == 1 or Path(parts[-1].lstrip("*")).name == zip_path.name:
                return candidate
    return None

def validate_environment_spec(
    actual_python,
    actual_packages,
    expected_python=EXPECTED_PYTHON_VERSION,
    expected_packages=EXPECTED_PACKAGE_VERSIONS,
):
    actual_python = str(actual_python)
    normalized_actual = {
        str(name): str(version)
        for name, version in actual_packages.items()
    }
    normalized_expected = {
        str(name): str(version)
        for name, version in expected_packages.items()
    }

    errors = []
    if actual_python != expected_python:
        errors.append(
            f"Python {actual_python} != expected {expected_python}"
        )

    for name in sorted(normalized_expected):
        actual = normalized_actual.get(name)
        expected = normalized_expected[name]
        if actual != expected:
            errors.append(
                f"{name} {actual!r} != expected {expected!r}"
            )

    if errors:
        raise RuntimeError(
            "Canonical Track A/B environment mismatch: "
            + " | ".join(errors)
        )

    return {
        "status": "PASS",
        "python": actual_python,
        "packages": {
            name: normalized_actual[name]
            for name in sorted(normalized_expected)
        },
    }


def verify_canonical_environment():
    actual_packages = {}
    for name in sorted(EXPECTED_PACKAGE_VERSIONS):
        try:
            actual_packages[name] = im.version(name)
        except Exception as exc:
            raise RuntimeError(
                f"Required canonical package unavailable: {name}: "
                f"{type(exc).__name__}: {exc}"
            )

    return validate_environment_spec(
        platform.python_version(),
        actual_packages,
    )


def deterministic_stage01_zip_sha(root, destination):
    root = Path(root).resolve()
    destination = Path(destination).resolve()

    if destination == root or root in destination.parents:
        raise RuntimeError(
            "Temporary reconstructed ZIP must be outside the frozen tree"
        )

    destination.parent.mkdir(parents=True, exist_ok=True)
    destination.unlink(missing_ok=True)

    files = sorted(
        [p for p in root.rglob("*") if p.is_file()],
        key=lambda p: p.relative_to(root).as_posix(),
    )

    with zipfile.ZipFile(
        destination,
        "w",
        compression=zipfile.ZIP_DEFLATED,
        compresslevel=6,
        allowZip64=True,
    ) as zf:
        for path in files:
            rel = normalize_tree_relpath(
                path.relative_to(root).as_posix()
            )
            arcname = f"{FREEZE_ID}/{rel}"
            info = zipfile.ZipInfo(
                arcname,
                date_time=(1980, 1, 1, 0, 0, 0),
            )
            info.create_system = 3
            info.compress_type = zipfile.ZIP_DEFLATED
            info.external_attr = 0o100644 << 16

            with open(path, "rb") as src, zf.open(
                info,
                "w",
                force_zip64=True,
            ) as dst:
                shutil.copyfileobj(
                    src,
                    dst,
                    length=8 * 1024 * 1024,
                )

    with zipfile.ZipFile(destination, "r") as zf:
        bad_member = zf.testzip()
        if bad_member is not None:
            raise RuntimeError(
                f"Reconstructed Stage-01 ZIP CRC failure: {bad_member}"
            )

    return {
        "reconstructed_zip_path": str(destination),
        "reconstructed_zip_sha256": sha256_file(destination),
        "member_count": len(files),
    }


def verify_extracted_root_against_canonical_zip(
    root,
    expected_zip_sha256=EXPECTED_ZIP_SHA256,
    temp_root=TEMP_ROOT,
):
    expected_zip_sha256 = validate_full_sha256(
        expected_zip_sha256,
        "expected canonical Stage-01 ZIP SHA-256",
    )

    root = Path(root).resolve()
    temp_root = Path(temp_root).resolve()
    root_key = hashlib.sha256(
        str(root).encode("utf-8")
    ).hexdigest()[:16]
    reconstructed = (
        temp_root
        / "stage01_zip_anchor"
        / f"{FREEZE_ID}-{root_key}.zip"
    )

    rec = deterministic_stage01_zip_sha(root, reconstructed)
    actual = rec["reconstructed_zip_sha256"]

    try:
        if actual != expected_zip_sha256:
            raise RuntimeError(
                "Extracted Stage-01 tree does not reconstruct to the "
                "canonical frozen ZIP identity: "
                f"{actual} != {expected_zip_sha256}"
            )
        return {
            "status": "PASS",
            "method": (
                "deterministic_reconstruction_matches_canonical_zip_sha256"
            ),
            "expected_zip_sha256": expected_zip_sha256,
            "verified_reconstructed_zip_sha256": actual,
            "member_count": rec["member_count"],
        }
    finally:
        reconstructed.unlink(missing_ok=True)

def verify_manifest_root(root):
    root = Path(root).resolve()

    for name in CONTROL_FILES:
        if not (root / name).is_file():
            raise RuntimeError(f"Missing control file: {name}")
    for name in REQUIRED_DATASET_DIRS:
        if not (root / name).is_dir():
            raise RuntimeError(f"Missing dataset directory: {name}")

    sums = parse_sha256sums(root / "SHA256SUMS.txt")
    verified_tree_members = verify_exact_tree_member_set(root, sums)

    verified_files = []
    for rel in sorted(sums):
        p = safe_child(root, rel)
        if not p.is_file():
            raise RuntimeError(f"SHA256SUMS path missing: {rel}")
        actual = sha256_file(p)
        if actual != sums[rel]:
            raise RuntimeError(
                f"SHA256SUMS mismatch for {rel}: {actual} != {sums[rel]}"
            )
        verified_files.append({
            "path": rel,
            "size_bytes": p.stat().st_size,
            "sha256": actual,
        })

    content_tree_fingerprint = canonical_tree_fingerprint_from_sums(sums)
    validate_full_sha256(
        content_tree_fingerprint,
        "computed content_tree_fingerprint",
    )

    manifest = read_json(root / "ACQUISITION_MANIFEST.json")
    if manifest.get("freeze_id") != FREEZE_ID:
        raise RuntimeError(
            f"Manifest freeze_id mismatch: {manifest.get('freeze_id')}"
        )
    if manifest.get("all_required_pass") is not True:
        raise RuntimeError("Manifest does not record all_required_pass=true")
    required = manifest.get("required_datasets", [])
    if set(required) != set(REQUIRED_DATASET_DIRS):
        raise RuntimeError(
            f"Manifest required_datasets mismatch: {required}"
        )

    manifest_datasets = manifest.get("datasets", {})
    if set(manifest_datasets) != set(REQUIRED_DATASET_DIRS):
        raise RuntimeError(
            "Manifest datasets key-set mismatch: "
            f"{sorted(manifest_datasets)}"
        )

    for dataset in sorted(manifest_datasets):
        drec = manifest_datasets[dataset]
        if drec.get("status") != "PASS":
            raise RuntimeError(
                f"Manifest dataset status is not PASS: {dataset}"
            )
        for item in drec.get("files", []):
            rel = normalize_tree_relpath(item["path"])
            p = safe_child(root, rel)
            if not p.is_file():
                raise RuntimeError(f"Manifest file missing: {rel}")
            if p.stat().st_size != int(item["size_bytes"]):
                raise RuntimeError(f"Manifest size mismatch: {rel}")
            actual = sha256_file(p)
            if actual != item["sha256"].lower():
                raise RuntimeError(f"Manifest SHA mismatch: {rel}")
            if rel in sums and sums[rel] != actual:
                raise RuntimeError(
                    f"Manifest/SHA256SUMS disagreement: {rel}"
                )

    status = read_json(root / "ACQUISITION_STATUS.json")
    for dataset in REQUIRED_DATASET_DIRS:
        if status.get("datasets", {}).get(dataset, {}).get("status") != "PASS":
            raise RuntimeError(
                f"ACQUISITION_STATUS is not PASS for {dataset}"
            )
    if status.get("package", {}).get("status") != "PASS":
        raise RuntimeError("ACQUISITION_STATUS package is not PASS")

    return {
        "root": str(root),
        "sha256sums": sums,
        "verified_files": verified_files,
        "verified_tree_members": verified_tree_members,
        "tree_fingerprint_algorithm": STAGE01_TREE_FINGERPRINT_ALGORITHM,
        "content_tree_fingerprint": content_tree_fingerprint,
        "sources_lock": read_json(root / "SOURCES_LOCK.json"),
        "manifest": manifest,
        "status": status,
        "sha256sums_file_sha256": sha256_file(root / "SHA256SUMS.txt"),
        "sources_lock_sha256": sha256_file(root / "SOURCES_LOCK.json"),
        "manifest_sha256": sha256_file(root / "ACQUISITION_MANIFEST.json"),
    }

def verify_source_identities(root_record):
    sources = root_record["sources_lock"].get("sources", {})
    checks = []

    def require_equal(dataset, field, actual, expected):
        ok = actual == expected
        checks.append({
            "dataset": dataset,
            "field": field,
            "actual": actual,
            "expected": expected,
            "pass": ok,
        })
        if not ok:
            raise RuntimeError(
                f"{dataset} source identity mismatch: {field}={actual!r}, expected {expected!r}"
            )

    bb = sources.get("BanglaBait", {})
    require_equal("BanglaBait", "repo_id", bb.get("repo_id"), EXPECTED_SOURCE_IDENTITIES["BanglaBait"]["repo_id"])
    require_equal("BanglaBait", "revision", bb.get("revision"), EXPECTED_SOURCE_IDENTITIES["BanglaBait"]["revision"])

    bc = sources.get("BanglaClick", {})
    require_equal("BanglaClick", "repo_url", bc.get("repo_url"), EXPECTED_SOURCE_IDENTITIES["BanglaClick"]["repo_url"])
    require_equal("BanglaClick", "commit", bc.get("commit"), EXPECTED_SOURCE_IDENTITIES["BanglaClick"]["commit"])

    bm = sources.get("BanMANI", {})
    require_equal("BanMANI", "repo_url", bm.get("repo_url"), EXPECTED_SOURCE_IDENTITIES["BanMANI"]["repo_url"])
    require_equal("BanMANI", "commit", bm.get("commit"), EXPECTED_SOURCE_IDENTITIES["BanMANI"]["commit"])

    bf = sources.get("BanFakeNews-2.0", {})
    auth = bf.get("authoritative", {})
    require_equal("BanFakeNews-2.0", "doi", auth.get("doi"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["doi"])
    require_equal("BanFakeNews-2.0", "version", auth.get("version"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["version"])
    main = bf.get("files", {}).get(EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["main_filename"], {})
    require_equal("BanFakeNews-2.0", "main_size", main.get("size"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["main_size"])
    require_equal("BanFakeNews-2.0", "main_sha256", main.get("sha256"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["main_sha256"])
    nt = bf.get("new_test", {})
    require_equal("BanFakeNews-2.0", "new_test_commit", nt.get("commit"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["new_test_commit"])
    require_equal("BanFakeNews-2.0", "new_test_sha256", nt.get("sha256"), EXPECTED_SOURCE_IDENTITIES["BanFakeNews-2.0"]["new_test_sha256"])

    bait = sources.get("BaitBuster-Bangla", {})
    auth = bait.get("authoritative", {})
    require_equal("BaitBuster-Bangla", "doi", auth.get("doi"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["doi"])
    require_equal("BaitBuster-Bangla", "version", auth.get("version"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["version"])
    ahf = bait.get("author_hf", {})
    require_equal("BaitBuster-Bangla", "hf_repo_id", ahf.get("repo_id"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["hf_repo_id"])
    require_equal("BaitBuster-Bangla", "hf_revision", ahf.get("revision"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["hf_revision"])
    canon = bait.get("canonical_research_artifact", {})
    require_equal("BaitBuster-Bangla", "canonical_filename", canon.get("filename"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["canonical_filename"])
    require_equal("BaitBuster-Bangla", "canonical_size", canon.get("size"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["canonical_size"])
    require_equal("BaitBuster-Bangla", "canonical_sha256", canon.get("sha256"), EXPECTED_SOURCE_IDENTITIES["BaitBuster-Bangla"]["canonical_sha256"])
    require_equal("BaitBuster-Bangla", "canonical_required_for_pass", canon.get("required_for_pass"), True)
    csv_rec = bait.get("publisher_listed_unavailable_serialization", {})
    require_equal("BaitBuster-Bangla", "csv_required_for_pass", csv_rec.get("required_for_pass"), False)
    require_equal("BaitBuster-Bangla", "csv_reconstruction_prohibited", csv_rec.get("reconstruction_prohibited"), True)

    return checks

def resolve_canonical_frozen_input(input_root):
    input_root = Path(input_root).resolve()

    zip_record = discover_canonical_zip_mount(
        input_root,
        required=False,
    )
    extracted_record = discover_canonical_extracted_root(
        input_root,
        required=False,
    )

    if zip_record is None and extracted_record is None:
        raise RuntimeError(
            "No canonical frozen input representation was found. "
            "Expected either the literal frozen ZIP or the already-extracted "
            f"{FREEZE_ID}/ root."
        )

    if zip_record is not None and extracted_record is not None:
        zip_verified = materialize_and_verify_canonical_zip(
            zip_record
        )
        zip_tree_fingerprint = zip_verified[
            "manifest_record"
        ]["content_tree_fingerprint"]
        extracted_tree_fingerprint = extracted_record[
            "manifest_record"
        ]["content_tree_fingerprint"]

        if zip_tree_fingerprint != extracted_tree_fingerprint:
            raise RuntimeError(
                "Literal-ZIP and extracted-root representations do not "
                "resolve to the same frozen tree identity."
            )

        if (
            extracted_record["tree_anchor"][
                "verified_reconstructed_zip_sha256"
            ]
            != EXPECTED_ZIP_SHA256
        ):
            raise RuntimeError(
                "Extracted-root representation failed canonical ZIP "
                "identity verification."
            )

        raise RuntimeError(
            "Ambiguous canonical frozen input: both a literal ZIP and an "
            "already-extracted root are mounted. Both resolve to the same "
            "frozen identity, but canonical execution requires exactly one "
            "representation. Remove one representation and rerun preflight."
        )

    if zip_record is not None:
        verified = materialize_and_verify_canonical_zip(
            zip_record
        )
        return {
            "input_mode": "ZIP_MODE",
            "input_representation": "ZIP",
            "root": verified["root"],
            "manifest_record": verified["manifest_record"],
            "frozen_input_mount_root":
                Path(zip_record["mount_root"]).resolve(),
            "zip_record": zip_record,
            "extracted_record": None,
        }

    return {
        "input_mode": "EXTRACTED_ROOT_MODE",
        "input_representation": "EXTRACTED_ROOT",
        "root": Path(extracted_record["root"]).resolve(),
        "manifest_record": extracted_record["manifest_record"],
        "frozen_input_mount_root":
            Path(extracted_record["mount_root"]).resolve(),
        "zip_record": None,
        "extracted_record": extracted_record,
    }

def discover_and_verify_input():
    canonical_input_root = Path("/kaggle/input")
    if not canonical_input_root.is_dir():
        raise RuntimeError(
            "Canonical Stage 02 execution requires Kaggle input root "
            "/kaggle/input."
        )

    # Runtime scientific input: frozen dataset only.
    selected = resolve_canonical_frozen_input(
        canonical_input_root
    )

    root = Path(selected["root"]).resolve()
    rec = selected["manifest_record"]
    frozen_mount_root = Path(
        selected["frozen_input_mount_root"]
    ).resolve()

    # Frozen source/version identities are part of preflight.
    source_identity_checks = verify_source_identities(
        rec
    )

    # Exact canonical runtime environment is checked before any scientific
    # artifact inspection, but it is independent of historical Notebook-01.
    environment_verification = verify_canonical_environment()

    if selected["input_mode"] == "ZIP_MODE":
        zip_record = selected["zip_record"]

        rec.update({
            "input_mode": "ZIP_MODE",
            "input_representation": "ZIP",
            "zip_path":
                str(Path(zip_record["zip_path"]).resolve()),
            "zip_sha256": EXPECTED_ZIP_SHA256,
            "zip_sidecar_path":
                str(Path(zip_record["zip_sidecar_path"]).resolve()),
            "zip_sha256_at_start":
                sha256_file(Path(zip_record["zip_path"])),
            "frozen_input_mount_root":
                str(frozen_mount_root),
            "environment_verification":
                environment_verification,
            "previous_stage02_outputs_used_as_input": False,
            "required_kaggle_dataset_inputs": 1,
            "physical_stage01_notebook_input_required": False,
            "tree_anchor": {
                "status": "PASS",
                "method": "verified_canonical_zip_sha256",
                "canonical_zip_sha256":
                    EXPECTED_ZIP_SHA256,
                "anchored_content_tree_fingerprint":
                    rec["content_tree_fingerprint"],
            },
        })

        log(
            "Accepted frozen input representation: ZIP mode — "
            + str(zip_record["zip_path"])
        )

    else:
        extracted_record = selected["extracted_record"]

        rec.update({
            "input_mode": "EXTRACTED_ROOT_MODE",
            "input_representation": "EXTRACTED_ROOT",
            "extracted_root_discovery": {
                "candidate_paths_examined":
                    extracted_record[
                        "candidate_paths_examined"
                    ],
                "candidate_evaluations":
                    extracted_record[
                        "candidate_evaluations"
                    ],
                "accepted_canonical_root_path":
                    extracted_record[
                        "accepted_canonical_root_path"
                    ],
                "canonical_verification_result":
                    extracted_record[
                        "canonical_verification_result"
                    ],
            },
            "zip_path": None,
            "zip_sha256": None,
            "zip_sidecar_path":
                str(
                    Path(
                        extracted_record["zip_sidecar_path"]
                    ).resolve()
                ),
            "zip_sha256_at_start": None,
            "frozen_input_mount_root":
                str(frozen_mount_root),
            "environment_verification":
                environment_verification,
            "previous_stage02_outputs_used_as_input": False,
            "required_kaggle_dataset_inputs": 1,
            "physical_stage01_notebook_input_required": False,
            "tree_anchor": {
                **extracted_record["tree_anchor"],
                "anchored_content_tree_fingerprint":
                    rec["content_tree_fingerprint"],
            },
        })

        log(
            "Accepted frozen input representation: extracted-root mode — "
            + str(root)
        )

    rec["source_identity_checks"] = (
        source_identity_checks
    )

    log(
        "Stage-01 Notebook SHA retained as historical provenance "
        "identifier only; no physical Notebook-01 runtime input is required."
    )

    return rec
````

## Mandatory Kaggle preflight gate

**Required Kaggle scientific input count: ONE.**

The only required runtime input is the frozen dataset beneath `/kaggle/input`.

A physical Stage-01 Notebook 01 is not required and is not part of preflight.
Previous Stage-02 outputs are not scientific inputs.

Before any dataset schema, missingness, duplicate, overlap, leakage, construct,
or split-firewall analysis is permitted, the preflight must:

1. resolve exactly one fully canonical frozen package representation/root;
2. verify the exact frozen ZIP identity directly or through deterministic
   reconstruction;
3. verify immutable frozen source/version/commit identities;
4. verify the complete required frozen package structure, manifest, status,
   SHA256SUMS, and tree membership;
5. verify the exact canonical Python/package environment;
6. set `P0_PREFLIGHT_PASS=True`.

Only after those checks pass may scientific artifact inspection begin.

No Kaggle-generated outer Dataset name is configured or trusted. No
user-editable input path is required.

The preflight log explicitly reports `ZIP_MODE` or `EXTRACTED_ROOT_MODE`.

Any failure raises immediately and terminates execution before scientific

## Schema, missingness, duplicate, and overlap methodology

### Protected BanglaBait scientific fields

BanglaBait train/dev/test use an explicit scientific-field contract:

- `title` = protected primary text;
- `label` = protected primary target.

The heuristic text-column selector remains available for descriptive auditing of external datasets and for identifying secondary BanglaBait text fields. It does not define the protected BanglaBait target field.

### Text fingerprints

For every selected text field, the audit computes:

- raw SHA-256 of the original string representation;
- normalized SHA-256 after Unicode NFKC normalization, zero-width removal, case folding, trimming, and whitespace collapse.

Punctuation is retained. Normalization is conservative and unchanged from P0-v1.0.0.

### Protected split leakage

Protected train/dev/test comparisons are field-attributed. The report contains a complete selected-field pair matrix.

The automatic leakage gate is evaluated only for:

`title` ↔ `title`

across train/dev/test.

Exact raw title overlap and normalized title overlap are separately counted. Normalization-only overlap is explicitly identified.

Secondary same-field and cross-field overlaps are retained as contamination evidence but do not masquerade as primary-title leakage. If a later Stage 03 protocol proposes to use a secondary field, that field must be brought under the firewall before any model training.

### Within-split conflicting labels

The protected conflict gate evaluates normalized `title` groups against `label`. It records:

- number of conflicting normalized-title groups;
- affected rows;
- label combinations;
- exact-raw versus normalization-only conflict status;
- malformed/placeholder status;
- short representative title excerpts.

No record is removed or relabeled.

### Diagnostic excerpts

Forensic output includes only short excerpts sufficient to identify the failure mode. Full article bodies are not written to the Stage 02 result directory.


````python

TEXT_POSITIVE = [
    "headline", "title", "text", "content", "article", "news",
    "caption", "description", "sentence", "statement", "transcript",
    "body", "summary", "teaser"
]
TEXT_NEGATIVE = [
    "label", "class", "target", "id", "url", "link", "date", "time",
    "source", "domain", "publisher", "author", "category", "index"
]
LABEL_POSITIVE = [
    "label", "class", "target", "clickbait", "fake", "manipulation",
    "manipulated", "is_fake", "is_clickbait", "bait"
]

def normalize_text(value):
    if value is None or (isinstance(value, float) and math.isnan(value)):
        return ""
    text = str(value)
    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u200b", "").replace("\u200c", "").replace("\u200d", "").replace("\ufeff", "")
    text = text.casefold().strip()
    text = re.sub(r"\s+", " ", text)
    return text

def sha_text(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

def column_name_score(name, positives, negatives=()):
    n = str(name).strip().lower()
    score = 0
    for token in positives:
        if n == token:
            score += 12
        elif token in n:
            score += 5
    for token in negatives:
        if token in n:
            score -= 8
    return score

def sample_artifact(path):
    suffix = path.suffix.lower()
    if suffix == ".csv":
        return pd.read_csv(path, nrows=500, encoding="utf-8-sig", low_memory=False)
    if suffix in {".xlsx", ".xls"}:
        return pd.read_excel(path, nrows=500)
    if suffix == ".parquet":
        pf = pq.ParquetFile(path)
        for batch in pf.iter_batches(batch_size=500):
            return batch.to_pandas()
        return pd.DataFrame(columns=pf.schema.names)
    raise RuntimeError(f"Unsupported artifact format: {path}")

def select_text_and_label_columns(sample):
    if sample is None or len(sample.columns) == 0:
        return [], [], None

    text_scored = []
    for col in sample.columns:
        s = sample[col]
        name_score = column_name_score(col, TEXT_POSITIVE, TEXT_NEGATIVE)
        if pd.api.types.is_string_dtype(s.dtype) or s.dtype == object:
            vals = s.dropna().astype(str).head(200)
            avg_len = float(vals.str.len().mean()) if len(vals) else 0.0
            length_score = min(avg_len / 20.0, 5.0)
            score = name_score + length_score
            text_scored.append((score, avg_len, str(col)))

    text_scored.sort(key=lambda x: (-x[0], -x[1], x[2].lower()))
    selected_text = [x[2] for x in text_scored if x[0] >= 4][:3]
    if not selected_text and text_scored and text_scored[0][1] >= 10:
        selected_text = [text_scored[0][2]]

    label_scored = []
    for col in sample.columns:
        s = sample[col]
        score = column_name_score(col, LABEL_POSITIVE, ())
        if score <= 0:
            continue
        vals = s.dropna()
        nunique_sample = int(vals.nunique(dropna=True)) if len(vals) else 0
        if nunique_sample <= 200:
            label_scored.append((score, nunique_sample, str(col)))
    label_scored.sort(key=lambda x: (-x[0], x[1], x[2].lower()))
    label_cols = [x[2] for x in label_scored[:5]]
    primary_label = label_cols[0] if label_cols else None
    return selected_text, label_cols, primary_label

def iter_chunks(path, preserve_physical_strings=False):
    suffix = path.suffix.lower()
    if suffix == ".csv":
        kwargs = {
            "encoding": "utf-8-sig",
            "low_memory": False,
            "chunksize": 25000,
        }
        if preserve_physical_strings:
            kwargs.update({
                "dtype": str,
                "keep_default_na": False,
                "na_filter": False,
            })
        yield from pd.read_csv(path, **kwargs)
        return
    if suffix in {".xlsx", ".xls"}:
        yield pd.read_excel(path)
        return
    if suffix == ".parquet":
        pf = pq.ParquetFile(path)
        for batch in pf.iter_batches(batch_size=25000):
            yield batch.to_pandas()
        return
    raise RuntimeError(f"Unsupported artifact format: {path}")

def simhash64(text, max_tokens=32):
    tokens = text.split()
    if not tokens:
        return 0
    tokens = tokens[:max_tokens]
    features = list(tokens)
    features += [tokens[i] + "\u241f" + tokens[i + 1] for i in range(len(tokens) - 1)]
    vector = [0] * 64
    for feature in features:
        hv = int.from_bytes(
            hashlib.blake2b(feature.encode("utf-8"), digest_size=8).digest(),
            "big",
        )
        for bit in range(64):
            vector[bit] += 1 if ((hv >> bit) & 1) else -1
    result = 0
    for bit, value in enumerate(vector):
        if value >= 0:
            result |= (1 << bit)
    return result

def hamming64(a, b):
    return (a ^ b).bit_count()

def inspect_reference_xlsx(path, artifact):
    wb = load_workbook(path, read_only=True, data_only=False)
    ws = wb[wb.sheetnames[0]]
    rows = ws.iter_rows(values_only=True)
    header = next(rows, tuple())
    approx_rows = max((ws.max_row or 1) - 1, 0)
    records = []
    for col in header:
        if col is None:
            continue
        records.append({
            "dataset": artifact["dataset"],
            "artifact_id": artifact["artifact_id"],
            "relative_path": artifact["relative_path"],
            "role": artifact["role"],
            "row_count": approx_rows,
            "column_name": str(col),
            "dtype_observed": "reference_xlsx_not_full_scanned",
            "non_null_count": None,
            "missing_count": None,
            "missing_pct": None,
            "primary_text": False,
            "primary_label": False,
            "inspection_scope": "secondary_reference_schema_header_only",
        })
    wb.close()
    return records

def initialize_fingerprint_db(db_path):
    db_path = Path(db_path)
    if db_path.exists():
        db_path.unlink()
    con = sqlite3.connect(db_path)
    con.execute("""
        CREATE TABLE fingerprints (
            dataset TEXT NOT NULL,
            artifact_id TEXT NOT NULL,
            role TEXT NOT NULL,
            protected_split TEXT,
            text_field TEXT NOT NULL,
            is_primary INTEGER NOT NULL,
            row_id INTEGER NOT NULL,
            raw_hash TEXT NOT NULL,
            norm_hash TEXT NOT NULL,
            norm_len INTEGER NOT NULL,
            label_raw TEXT,
            label TEXT,
            title_quality TEXT,
            raw_text_protected TEXT,
            norm_text_protected TEXT
        )
    """)
    con.execute("CREATE INDEX idx_fp_artifact ON fingerprints(artifact_id)")
    con.execute("CREATE INDEX idx_fp_raw ON fingerprints(raw_hash)")
    con.execute("CREATE INDEX idx_fp_norm ON fingerprints(norm_hash)")
    con.execute("CREATE INDEX idx_fp_artifact_norm ON fingerprints(artifact_id, norm_hash)")
    con.execute(
        "CREATE INDEX idx_fp_field_norm "
        "ON fingerprints(artifact_id, text_field, norm_hash, row_id)"
    )
    return con

def audit_artifact(root, artifact, con):
    path = root / artifact["relative_path"]
    if not path.is_file():
        if artifact["artifact_id"] == "baitbuster_xlsx_reference":
            add_finding(
                "MINOR",
                "P0-PROV-BAIT-XLSX",
                "provenance",
                "Secondary BaitBuster XLSX reference artifact is absent; canonical Parquet remains the required artifact.",
                False,
                {"path": artifact["relative_path"]},
            )
            return [], [], [], None
        raise FileNotFoundError(path)

    if not artifact["analyze_content"]:
        return inspect_reference_xlsx(path, artifact), [], [], {
            "artifact_id": artifact["artifact_id"],
            "dataset": artifact["dataset"],
            "role": artifact["role"],
            "relative_path": artifact["relative_path"],
            "row_count": None,
            "selected_text_columns": [],
            "label_columns": [],
            "primary_label": None,
            "inspection_scope": "secondary_reference_schema_header_only",
        }

    sample = sample_artifact(path)
    heuristic_selected_text, label_cols, heuristic_primary_label = (
        select_text_and_label_columns(sample)
    )

    selected_text = list(heuristic_selected_text)
    primary_label = heuristic_primary_label
    primary_text_column = selected_text[0] if selected_text else None

    protected_primary_text = artifact.get("protected_primary_text_column")
    protected_primary_label = artifact.get("protected_primary_label_column")
    protected_split = artifact.get("protected_split") in {
        "train",
        "calibration",
        "test",
    }

    protected_raw = None
    protected_raw_consistency = None
    present_protected_text_fields = []
    absent_protected_text_fields = []

    if protected_primary_text:
        if protected_primary_text not in sample.columns:
            raise RuntimeError(
                f"Required protected primary text column {protected_primary_text!r} "
                f"is absent from {artifact['artifact_id']}; columns={list(sample.columns)}"
            )
        if path.suffix.lower() != ".csv":
            raise RuntimeError(
                "Protected BanglaBait forensic path requires physical CSV input"
            )

        present_protected_text_fields = [
            field
            for field in PROTECTED_AUDITED_TEXT_FIELDS
            if field in sample.columns
        ]
        absent_protected_text_fields = [
            field
            for field in PROTECTED_AUDITED_TEXT_FIELDS
            if field not in sample.columns
        ]
        selected_text = list(present_protected_text_fields)
        primary_text_column = protected_primary_text

        protected_raw = load_protected_csv_raw_records(path, artifact)
        protected_raw_consistency = (
            verify_raw_vs_pandas_protected_consistency(
                path,
                protected_raw,
            )
        )

    if protected_primary_label:
        if protected_primary_label not in sample.columns:
            raise RuntimeError(
                f"Required protected primary label column {protected_primary_label!r} "
                f"is absent from {artifact['artifact_id']}; columns={list(sample.columns)}"
            )
        primary_label = protected_primary_label
        if protected_primary_label not in label_cols:
            label_cols = [protected_primary_label] + label_cols

    if artifact["requires_primary_text"] and not selected_text:
        raise RuntimeError(
            f"No usable primary text column identified for {artifact['artifact_id']}; "
            f"columns={list(sample.columns)}"
        )
    if artifact["requires_primary_label"] and primary_label is None:
        raise RuntimeError(
            f"No usable label column identified for {artifact['artifact_id']}; "
            f"columns={list(sample.columns)}"
        )

    schema_acc = {}
    label_counts = {col: Counter() for col in label_cols}
    row_count = 0
    text_nonempty = Counter()
    primary_label_missing = 0
    primary_label_values = Counter()

    insert_buffer = []
    MIN_CHARS = int(AUDIT_CONFIG["text_fingerprint_min_chars"])

    def flush_insert_buffer():
        if insert_buffer:
            con.executemany(
                "INSERT INTO fingerprints VALUES (?,?,?,?,?,?,?,?,?,?,?,?,?,?,?)",
                insert_buffer,
            )
            con.commit()
            insert_buffer.clear()

    # Protected title fingerprints are derived from the literal-preserving
    # physical CSV pass. Every non-empty protected title is fingerprinted,
    # including valid short titles and non-empty malformed/placeholder titles.
    if protected_raw is not None:
        for rec in protected_raw["records"]:
            raw_title = rec["raw_title"]
            if not protected_title_should_fingerprint(raw_title):
                continue
            norm = rec["normalized_title"]
            text_nonempty[protected_primary_text] += 1
            insert_buffer.append((
                artifact["dataset"],
                artifact["artifact_id"],
                artifact["role"],
                artifact.get("protected_split"),
                protected_primary_text,
                1,
                rec["row_id"],
                sha_text(str(raw_title)),
                sha_text(norm),
                len(norm),
                rec["raw_label"],
                rec["canonical_label"],
                rec["title_quality"],
                truncate_diagnostic_text(raw_title),
                norm,
            ))
            if len(insert_buffer) >= 10000:
                flush_insert_buffer()
        flush_insert_buffer()

    for chunk in iter_chunks(path):
        chunk.columns = [str(c) for c in chunk.columns]
        n = len(chunk)
        row_start = row_count
        row_count += n

        for col in chunk.columns:
            s = chunk[col]
            rec = schema_acc.setdefault(
                col,
                {
                    "types": set(),
                    "non_null": 0,
                    "missing": 0,
                },
            )
            rec["types"].add(str(s.dtype))
            rec["non_null"] += int(s.notna().sum())
            rec["missing"] += int(s.isna().sum())

        for col in label_cols:
            if col not in chunk.columns:
                continue
            counts = (
                chunk[col]
                .astype("string")
                .fillna("<MISSING>")
                .value_counts(dropna=False)
            )
            for value, count in counts.items():
                label_counts[col][str(value)] += int(count)

        if protected_raw is None:
            if primary_label and primary_label in chunk.columns:
                lab_series = chunk[primary_label].astype("string")
                primary_label_missing += int(lab_series.isna().sum())
                for value, count in (
                    lab_series.fillna("<MISSING>").value_counts().items()
                ):
                    primary_label_values[str(value)] += int(count)
            else:
                lab_series = pd.Series(
                    [pd.NA] * n,
                    index=chunk.index,
                    dtype="string",
                )
        else:
            lab_series = None

        for text_col in selected_text:
            if text_col not in chunk.columns:
                continue

            # Protected primary title has already been inserted from the raw
            # literal-preserving CSV pass.
            if protected_raw is not None and text_col == protected_primary_text:
                continue

            vals = chunk[text_col]
            for offset, value in enumerate(vals.tolist()):
                if pd.isna(value):
                    continue

                raw = str(value)
                norm = normalize_text(raw)

                is_protected_audited_field = (
                    protected_split
                    and text_col in PROTECTED_AUDITED_TEXT_FIELDS
                )
                if is_protected_audited_field:
                    if norm == "":
                        continue
                elif len(norm) < MIN_CHARS:
                    continue

                text_nonempty[text_col] += 1

                if protected_raw is not None:
                    raw_rec = protected_raw["records"][row_start + offset]
                    label_raw = raw_rec["raw_label"]
                    label_value = raw_rec["canonical_label"]
                else:
                    label_raw = None
                    label_value = None
                    if primary_label and primary_label in chunk.columns:
                        lv = lab_series.iloc[offset]
                        if pd.notna(lv):
                            label_raw = str(lv)
                            label_value = str(lv)

                is_primary = int(text_col == primary_text_column)
                protected_excerpt = (
                    truncate_diagnostic_text(raw)
                    if protected_split
                    else None
                )
                protected_norm_primary = (
                    norm
                    if protected_split and is_primary
                    else None
                )

                insert_buffer.append((
                    artifact["dataset"],
                    artifact["artifact_id"],
                    artifact["role"],
                    artifact.get("protected_split"),
                    text_col,
                    is_primary,
                    row_start + offset,
                    sha_text(raw),
                    sha_text(norm),
                    len(norm),
                    label_raw,
                    label_value,
                    None,
                    protected_excerpt,
                    protected_norm_primary,
                ))
                if len(insert_buffer) >= 10000:
                    flush_insert_buffer()

    flush_insert_buffer()

    if protected_raw is not None:
        if row_count != protected_raw["row_count"]:
            raise RuntimeError(
                "Existing audit reader row-count mismatch against protected "
                f"raw pass: pandas={row_count}; raw={protected_raw['row_count']}"
            )
        primary_label_missing = sum(
            1
            for rec in protected_raw["records"]
            if rec["canonical_label"] == ""
        )
        primary_label_values = Counter(
            rec["canonical_label"]
            for rec in protected_raw["records"]
            if rec["canonical_label"] != ""
        )

    schema_rows = []
    for col in sorted(schema_acc):
        rec = schema_acc[col]
        schema_rows.append({
            "dataset": artifact["dataset"],
            "artifact_id": artifact["artifact_id"],
            "relative_path": artifact["relative_path"],
            "role": artifact["role"],
            "row_count": row_count,
            "column_name": col,
            "dtype_observed": "|".join(sorted(rec["types"])),
            "non_null_count": rec["non_null"],
            "missing_count": rec["missing"],
            "missing_pct": (rec["missing"] / row_count) if row_count else None,
            "primary_text": bool(
                primary_text_column and col == primary_text_column
            ),
            "selected_text": col in selected_text,
            "primary_label": col == primary_label,
            "label_like": col in label_cols,
            "inspection_scope": "full",
        })

    label_rows = []
    for col in sorted(label_cols):
        total = sum(label_counts[col].values())
        for value in sorted(label_counts[col]):
            count = label_counts[col][value]
            label_rows.append({
                "dataset": artifact["dataset"],
                "artifact_id": artifact["artifact_id"],
                "role": artifact["role"],
                "label_column": col,
                "label_value": value,
                "count": count,
                "proportion": count / total if total else None,
                "primary_label": col == primary_label,
            })

    summary = {
        "artifact_id": artifact["artifact_id"],
        "dataset": artifact["dataset"],
        "role": artifact["role"],
        "relative_path": artifact["relative_path"],
        "row_count": row_count,
        "selected_text_columns": list(selected_text),
        "heuristic_selected_text_columns": list(heuristic_selected_text),
        "primary_text_column": primary_text_column,
        "label_columns": sorted(set(label_cols)),
        "primary_label": primary_label,
        "primary_label_missing": primary_label_missing,
        "primary_label_class_count": len(primary_label_values),
        "primary_text_eligible_rows": (
            text_nonempty.get(primary_text_column, 0)
            if primary_text_column
            else 0
        ),
        "present_protected_text_fields": list(
            present_protected_text_fields
        ),
        "absent_protected_text_fields": list(
            absent_protected_text_fields
        ),
        "protected_title_quality": (
            protected_raw["title_quality"]
            if protected_raw is not None
            else None
        ),
        "protected_label_domain": (
            protected_raw["label_domain"]
            if protected_raw is not None
            else None
        ),
        "raw_vs_pandas_consistency": protected_raw_consistency,
        "inspection_scope": "full",
    }
    return schema_rows, label_rows, [], summary

def duplicate_report(con, artifact_summaries):
    rows = []
    for summary in artifact_summaries:
        if summary is None or summary.get("inspection_scope") != "full":
            continue
        artifact_id = summary["artifact_id"]
        fields = [
            r[0] for r in con.execute(
                "SELECT DISTINCT text_field FROM fingerprints WHERE artifact_id=? ORDER BY text_field",
                (artifact_id,),
            ).fetchall()
        ]
        for field in fields:
            total = con.execute(
                "SELECT COUNT(*) FROM fingerprints WHERE artifact_id=? AND text_field=?",
                (artifact_id, field),
            ).fetchone()[0]
            unique_raw = con.execute(
                "SELECT COUNT(DISTINCT raw_hash) FROM fingerprints WHERE artifact_id=? AND text_field=?",
                (artifact_id, field),
            ).fetchone()[0]
            unique_norm = con.execute(
                "SELECT COUNT(DISTINCT norm_hash) FROM fingerprints WHERE artifact_id=? AND text_field=?",
                (artifact_id, field),
            ).fetchone()[0]
            raw_groups = con.execute(
                """SELECT COUNT(*) FROM (
                    SELECT raw_hash FROM fingerprints
                    WHERE artifact_id=? AND text_field=?
                    GROUP BY raw_hash HAVING COUNT(*)>1
                )""",
                (artifact_id, field),
            ).fetchone()[0]
            norm_groups = con.execute(
                """SELECT COUNT(*) FROM (
                    SELECT norm_hash FROM fingerprints
                    WHERE artifact_id=? AND text_field=?
                    GROUP BY norm_hash HAVING COUNT(*)>1
                )""",
                (artifact_id, field),
            ).fetchone()[0]
            conflict_groups = con.execute(
                """SELECT COUNT(*) FROM (
                    SELECT norm_hash FROM fingerprints
                    WHERE artifact_id=? AND text_field=? AND label IS NOT NULL
                    GROUP BY norm_hash HAVING COUNT(DISTINCT label)>1
                )""",
                (artifact_id, field),
            ).fetchone()[0]
            max_group = con.execute(
                """SELECT COALESCE(MAX(n),0) FROM (
                    SELECT COUNT(*) AS n FROM fingerprints
                    WHERE artifact_id=? AND text_field=?
                    GROUP BY norm_hash
                )""",
                (artifact_id, field),
            ).fetchone()[0]
            rows.append({
                "dataset": summary["dataset"],
                "artifact_id": artifact_id,
                "role": summary["role"],
                "text_field": field,
                "eligible_text_rows": total,
                "unique_raw_hashes": unique_raw,
                "raw_duplicate_rows": total - unique_raw,
                "raw_duplicate_groups": raw_groups,
                "unique_normalized_hashes": unique_norm,
                "normalized_duplicate_rows": total - unique_norm,
                "normalized_duplicate_groups": norm_groups,
                "conflicting_label_normalized_groups": conflict_groups,
                "max_normalized_group_size": max_group,
                "normalized_duplicate_rate": ((total - unique_norm) / total) if total else 0.0,
            })
    return rows


def artifact_pair_overlap_union(con, left, right):
    """Descriptive union-of-selected-fields overlap; never a protected-field gate."""
    exact_shared = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.raw_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.raw_hash=b.raw_hash
            WHERE a.artifact_id=? AND b.artifact_id=?
        )""",
        (left, right),
    ).fetchone()[0]
    norm_shared = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.norm_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.norm_hash=b.norm_hash
            WHERE a.artifact_id=? AND b.artifact_id=?
        )""",
        (left, right),
    ).fetchone()[0]
    left_rows = con.execute(
        """SELECT COUNT(DISTINCT a.row_id)
           FROM fingerprints a
           WHERE a.artifact_id=? AND EXISTS (
             SELECT 1 FROM fingerprints b
             WHERE b.artifact_id=? AND b.norm_hash=a.norm_hash
           )""",
        (left, right),
    ).fetchone()[0]
    right_rows = con.execute(
        """SELECT COUNT(DISTINCT b.row_id)
           FROM fingerprints b
           WHERE b.artifact_id=? AND EXISTS (
             SELECT 1 FROM fingerprints a
             WHERE a.artifact_id=? AND a.norm_hash=b.norm_hash
           )""",
        (right, left),
    ).fetchone()[0]
    conflict_hashes = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.norm_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.norm_hash=b.norm_hash
            WHERE a.artifact_id=? AND b.artifact_id=?
              AND a.label IS NOT NULL AND b.label IS NOT NULL
              AND a.label<>b.label
        )""",
        (left, right),
    ).fetchone()[0]
    return exact_shared, norm_shared, left_rows, right_rows, conflict_hashes


def field_semantic_role(field):
    name = str(field).strip().lower()
    if name == "title":
        return "headline/title"
    if name == "content":
        return "article/body"
    if name == "translated_title":
        return "translated_headline/title"
    if name == "translated_content":
        return "translated_article/body"
    if "url" in name or "link" in name or name == "domain":
        return "source/url/metadata"
    return "secondary_text_or_metadata"


def truncate_diagnostic_text(value, max_chars=None):
    if value is None:
        return None
    max_chars = max_chars or int(
        AUDIT_CONFIG["protected_field_policy"]["forensic_excerpt_max_chars"]
    )
    text = str(value).replace("\n", " ").replace("\r", " ")
    text = re.sub(r"\s+", " ", text).strip()
    return text if len(text) <= max_chars else text[:max_chars] + "…"


def canonicalize_label_for_comparison(raw_label):
    if raw_label is None:
        return ""
    return unicodedata.normalize("NFKC", str(raw_label)).strip()


def classify_protected_title(raw_title):
    if raw_title is None:
        return "MISSING"

    norm = normalize_text(raw_title)
    if norm == "":
        return "EMPTY"

    placeholders = set(
        AUDIT_CONFIG["protected_field_policy"][
            "known_placeholder_normalized_values"
        ]
    )
    if norm in placeholders:
        return "KNOWN_PLACEHOLDER"

    if not any(ch.isalnum() for ch in norm):
        return "NON_TEXTUAL_GARBAGE"

    if len(norm) < int(AUDIT_CONFIG["text_fingerprint_min_chars"]):
        return "VALID_SHORT"

    return "VALID"


def protected_title_should_fingerprint(raw_title):
    if raw_title is None:
        return False
    return normalize_text(raw_title) != ""


def _record_identity_cell(value):
    if value is None:
        return ""
    try:
        if pd.isna(value):
            return ""
    except Exception:
        pass
    return str(value)


def protected_record_identity(row, headers):
    values = [_record_identity_cell(row.get(col)) for col in headers]
    payload = json.dumps(
        values,
        ensure_ascii=False,
        separators=(",", ":"),
    ).encode("utf-8")
    return hashlib.sha256(payload).hexdigest()


def load_protected_csv_raw_records(path, artifact):
    title_col = artifact["protected_primary_text_column"]
    label_col = artifact["protected_primary_label_column"]

    records = []
    quality_counts = Counter()
    observed_label_counts = Counter()
    invalid_label_rows = []
    malformed_rows = []

    with open(path, "r", encoding="utf-8-sig", newline="") as f:
        reader = csv.DictReader(f)
        headers = [str(x) for x in (reader.fieldnames or [])]
        if title_col not in headers:
            raise RuntimeError(
                f"Protected title column {title_col!r} absent from raw CSV"
            )
        if label_col not in headers:
            raise RuntimeError(
                f"Protected label column {label_col!r} absent from raw CSV"
            )

        for row_id, row in enumerate(reader):
            raw_title = row.get(title_col)
            raw_label = row.get(label_col)
            canonical_label = canonicalize_label_for_comparison(raw_label)
            title_quality = classify_protected_title(raw_title)
            normalized_title = (
                normalize_text(raw_title)
                if raw_title is not None
                else ""
            )

            quality_counts[title_quality] += 1
            observed_label_counts[canonical_label] += 1

            if canonical_label not in EXPECTED_BANGLABAIT_LABEL_DOMAIN:
                invalid_label_rows.append({
                    "row_id": row_id,
                    "raw_label": raw_label,
                    "canonical_label": canonical_label,
                })

            if title_quality in {
                "MISSING",
                "EMPTY",
                "KNOWN_PLACEHOLDER",
                "NON_TEXTUAL_GARBAGE",
            }:
                malformed_rows.append({
                    "row_id": row_id,
                    "category": title_quality,
                    "raw_sha256": (
                        sha_text(str(raw_title))
                        if raw_title is not None
                        else None
                    ),
                    "normalized_sha256": (
                        sha_text(normalized_title)
                        if normalized_title
                        else None
                    ),
                    "raw_label": raw_label,
                    "canonical_label": canonical_label,
                    "excerpt": truncate_diagnostic_text(raw_title),
                })

            records.append({
                "row_id": row_id,
                "row_identity_sha256": protected_record_identity(row, headers),
                "raw_title": raw_title,
                "normalized_title": normalized_title,
                "title_quality": title_quality,
                "raw_label": raw_label,
                "canonical_label": canonical_label,
            })

    malformed_rows = sorted(
        malformed_rows,
        key=lambda x: (x["row_id"], x["category"]),
    )
    quality_counts_sorted = {
        key: int(quality_counts.get(key, 0))
        for key in [
            "MISSING",
            "EMPTY",
            "KNOWN_PLACEHOLDER",
            "NON_TEXTUAL_GARBAGE",
            "VALID_SHORT",
            "VALID",
        ]
    }
    observed_label_counts_sorted = {
        key: int(observed_label_counts[key])
        for key in sorted(observed_label_counts)
    }

    return {
        "headers": headers,
        "records": records,
        "row_count": len(records),
        "title_quality": {
            "row_count": len(records),
            "quality_counts": quality_counts_sorted,
            "malformed_count": sum(
                quality_counts_sorted[k]
                for k in [
                    "MISSING",
                    "EMPTY",
                    "KNOWN_PLACEHOLDER",
                    "NON_TEXTUAL_GARBAGE",
                ]
            ),
            "valid_short_count": quality_counts_sorted["VALID_SHORT"],
            "malformed_row_ids": [
                item["row_id"] for item in malformed_rows
            ],
            "representative_malformed_rows": malformed_rows[:12],
        },
        "label_domain": {
            "expected_canonical_domain": list(EXPECTED_BANGLABAIT_LABEL_DOMAIN),
            "observed_canonical_counts": observed_label_counts_sorted,
            "invalid_row_count": len(invalid_label_rows),
            "invalid_rows": sorted(
                invalid_label_rows,
                key=lambda x: x["row_id"],
            ),
            "pass": len(invalid_label_rows) == 0,
            "canonicalization": "Unicode NFKC + surrounding-whitespace trim only; no semantic mapping",
        },
    }


def verify_raw_vs_pandas_protected_consistency(path, raw_result):
    expected_headers = list(raw_result["headers"])
    raw_records = raw_result["records"]
    observed_rows = 0

    for chunk in iter_chunks(path, preserve_physical_strings=True):
        chunk.columns = [str(c) for c in chunk.columns]
        if list(chunk.columns) != expected_headers:
            raise RuntimeError(
                "Raw-vs-pandas protected header/order mismatch: "
                f"raw={expected_headers}; pandas={list(chunk.columns)}"
            )

        for _, series in chunk.iterrows():
            if observed_rows >= len(raw_records):
                raise RuntimeError(
                    "Pandas protected reader produced more rows than raw CSV reader"
                )
            pandas_row = {
                col: series[col]
                for col in expected_headers
            }
            observed_identity = protected_record_identity(
                pandas_row,
                expected_headers,
            )
            expected_identity = raw_records[observed_rows][
                "row_identity_sha256"
            ]
            if observed_identity != expected_identity:
                raise RuntimeError(
                    "Raw-vs-pandas protected row-identity mismatch at row "
                    f"{observed_rows}: "
                    f"{observed_identity} != {expected_identity}"
                )
            observed_rows += 1

    if observed_rows != len(raw_records):
        raise RuntimeError(
            "Raw-vs-pandas protected row-count mismatch: "
            f"pandas={observed_rows}; raw={len(raw_records)}"
        )

    return {
        "status": "PASS",
        "raw_row_count": len(raw_records),
        "pandas_row_count": observed_rows,
        "identity_fields": list(expected_headers),
        "ordering_verified": True,
        "record_identity_verified": True,
        "pandas_mode": "dtype=str, keep_default_na=False, na_filter=False",
    }


def field_pair_overlap(
    con,
    left,
    right,
    left_field,
    right_field,
    sample_limit=4,
):
    params = (left, left_field, right, right_field)

    exact_shared = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.raw_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.raw_hash=b.raw_hash
            WHERE a.artifact_id=? AND a.text_field=?
              AND b.artifact_id=? AND b.text_field=?
        )""",
        params,
    ).fetchone()[0]

    norm_shared = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.norm_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.norm_hash=b.norm_hash
            WHERE a.artifact_id=? AND a.text_field=?
              AND b.artifact_id=? AND b.text_field=?
        )""",
        params,
    ).fetchone()[0]

    def affected_rows(side, hash_field):
        if side == "left":
            return con.execute(
                f"""SELECT COUNT(DISTINCT a.row_id)
                    FROM fingerprints a
                    WHERE a.artifact_id=? AND a.text_field=?
                      AND EXISTS (
                        SELECT 1 FROM fingerprints b
                        WHERE b.artifact_id=? AND b.text_field=?
                          AND b.{hash_field}=a.{hash_field}
                      )""",
                params,
            ).fetchone()[0]
        return con.execute(
            f"""SELECT COUNT(DISTINCT b.row_id)
                FROM fingerprints b
                WHERE b.artifact_id=? AND b.text_field=?
                  AND EXISTS (
                    SELECT 1 FROM fingerprints a
                    WHERE a.artifact_id=? AND a.text_field=?
                      AND a.{hash_field}=b.{hash_field}
                  )""",
            (right, right_field, left, left_field),
        ).fetchone()[0]

    exact_left_rows = affected_rows("left", "raw_hash")
    exact_right_rows = affected_rows("right", "raw_hash")
    normalized_left_rows = affected_rows("left", "norm_hash")
    normalized_right_rows = affected_rows("right", "norm_hash")

    normalization_only_left_rows = con.execute(
        """SELECT COUNT(DISTINCT a.row_id)
           FROM fingerprints a
           WHERE a.artifact_id=? AND a.text_field=?
             AND EXISTS (
               SELECT 1 FROM fingerprints b
               WHERE b.artifact_id=? AND b.text_field=?
                 AND b.norm_hash=a.norm_hash
             )
             AND NOT EXISTS (
               SELECT 1 FROM fingerprints b
               WHERE b.artifact_id=? AND b.text_field=?
                 AND b.raw_hash=a.raw_hash
             )""",
        params + (right, right_field),
    ).fetchone()[0]

    normalization_only_right_rows = con.execute(
        """SELECT COUNT(DISTINCT b.row_id)
           FROM fingerprints b
           WHERE b.artifact_id=? AND b.text_field=?
             AND EXISTS (
               SELECT 1 FROM fingerprints a
               WHERE a.artifact_id=? AND a.text_field=?
                 AND a.norm_hash=b.norm_hash
             )
             AND NOT EXISTS (
               SELECT 1 FROM fingerprints a
               WHERE a.artifact_id=? AND a.text_field=?
                 AND a.raw_hash=b.raw_hash
             )""",
        (right, right_field, left, left_field, left, left_field),
    ).fetchone()[0]

    conflict_hashes = con.execute(
        """SELECT COUNT(*) FROM (
            SELECT DISTINCT a.norm_hash
            FROM fingerprints a
            JOIN fingerprints b ON a.norm_hash=b.norm_hash
            WHERE a.artifact_id=? AND a.text_field=?
              AND b.artifact_id=? AND b.text_field=?
              AND a.label IS NOT NULL AND b.label IS NOT NULL
              AND a.label<>b.label
        )""",
        params,
    ).fetchone()[0]

    normalization_only_shared = con.execute(
        """WITH shared_norm AS (
               SELECT DISTINCT a.norm_hash AS h
               FROM fingerprints a
               JOIN fingerprints b ON a.norm_hash=b.norm_hash
               WHERE a.artifact_id=? AND a.text_field=?
                 AND b.artifact_id=? AND b.text_field=?
           ),
           exact_norm AS (
               SELECT DISTINCT a.norm_hash AS h
               FROM fingerprints a
               JOIN fingerprints b ON a.raw_hash=b.raw_hash
               WHERE a.artifact_id=? AND a.text_field=?
                 AND b.artifact_id=? AND b.text_field=?
           )
           SELECT COUNT(*)
           FROM shared_norm
           WHERE h NOT IN (SELECT h FROM exact_norm)""",
        params + params,
    ).fetchone()[0]

    label_pair_rows = con.execute(
        """SELECT a.label, b.label, COUNT(DISTINCT a.norm_hash)
           FROM fingerprints a
           JOIN fingerprints b ON a.norm_hash=b.norm_hash
           WHERE a.artifact_id=? AND a.text_field=?
             AND b.artifact_id=? AND b.text_field=?
             AND a.label IS NOT NULL AND b.label IS NOT NULL
           GROUP BY a.label, b.label
           ORDER BY a.label, b.label""",
        params,
    ).fetchall()
    label_pair_shared_hash_counts = [
        {
            "left_label": row[0],
            "right_label": row[1],
            "shared_normalized_hashes": int(row[2]),
        }
        for row in label_pair_rows
    ]

    def overlapping_quality_counts(
        artifact_id,
        text_field,
        other_artifact,
        other_field,
    ):
        rows = con.execute(
            """SELECT COALESCE(a.title_quality, '<NONE>'), COUNT(DISTINCT a.row_id)
               FROM fingerprints a
               WHERE a.artifact_id=? AND a.text_field=?
                 AND EXISTS (
                   SELECT 1 FROM fingerprints b
                   WHERE b.artifact_id=? AND b.text_field=?
                     AND b.norm_hash=a.norm_hash
                 )
               GROUP BY COALESCE(a.title_quality, '<NONE>')
               ORDER BY COALESCE(a.title_quality, '<NONE>')""",
            (artifact_id, text_field, other_artifact, other_field),
        ).fetchall()
        return {
            str(category): int(count)
            for category, count in rows
        }

    left_quality_counts = overlapping_quality_counts(
        left,
        left_field,
        right,
        right_field,
    )
    right_quality_counts = overlapping_quality_counts(
        right,
        right_field,
        left,
        left_field,
    )

    shared_hashes = [
        r[0] for r in con.execute(
            """SELECT DISTINCT a.norm_hash
               FROM fingerprints a
               JOIN fingerprints b ON a.norm_hash=b.norm_hash
               WHERE a.artifact_id=? AND a.text_field=?
                 AND b.artifact_id=? AND b.text_field=?
               ORDER BY a.norm_hash
               LIMIT ?""",
            params + (sample_limit,),
        ).fetchall()
    ]

    samples = []
    for norm_hash in shared_hashes:
        exact_exists = bool(
            con.execute(
                """SELECT EXISTS(
                     SELECT 1
                     FROM fingerprints a
                     JOIN fingerprints b ON a.raw_hash=b.raw_hash
                     WHERE a.artifact_id=? AND a.text_field=?
                       AND b.artifact_id=? AND b.text_field=?
                       AND a.norm_hash=? AND b.norm_hash=?
                   )""",
                params + (norm_hash, norm_hash),
            ).fetchone()[0]
        )

        left_examples = con.execute(
            """SELECT row_id, raw_hash, label_raw, label,
                      title_quality, raw_text_protected
               FROM fingerprints
               WHERE artifact_id=? AND text_field=? AND norm_hash=?
               ORDER BY row_id, raw_hash
               LIMIT 3""",
            (left, left_field, norm_hash),
        ).fetchall()
        right_examples = con.execute(
            """SELECT row_id, raw_hash, label_raw, label,
                      title_quality, raw_text_protected
               FROM fingerprints
               WHERE artifact_id=? AND text_field=? AND norm_hash=?
               ORDER BY row_id, raw_hash
               LIMIT 3""",
            (right, right_field, norm_hash),
        ).fetchall()

        def render_examples(rows):
            return [
                {
                    "row_id": r[0],
                    "raw_sha256": r[1],
                    "raw_label": r[2],
                    "canonical_label": r[3],
                    "title_quality": r[4],
                    "excerpt": truncate_diagnostic_text(r[5]),
                }
                for r in rows
            ]

        samples.append({
            "normalized_sha256": norm_hash,
            "exact_raw_overlap_present": exact_exists,
            "left": render_examples(left_examples),
            "right": render_examples(right_examples),
        })

    return {
        "left_artifact": left,
        "right_artifact": right,
        "left_text_field": left_field,
        "right_text_field": right_field,
        "left_field_role": field_semantic_role(left_field),
        "right_field_role": field_semantic_role(right_field),
        "same_field": left_field == right_field,
        "exact_raw_shared_hashes": int(exact_shared),
        "normalized_shared_hashes": int(norm_shared),
        "normalization_only_shared_hashes": int(
            normalization_only_shared
        ),
        "exact_raw_affected_rows_left": int(exact_left_rows),
        "exact_raw_affected_rows_right": int(exact_right_rows),
        "normalized_affected_rows_left": int(normalized_left_rows),
        "normalized_affected_rows_right": int(normalized_right_rows),
        "normalization_only_affected_rows_left": int(
            normalization_only_left_rows
        ),
        "normalization_only_affected_rows_right": int(
            normalization_only_right_rows
        ),
        "conflicting_label_shared_hashes": int(conflict_hashes),
        "label_pair_shared_hash_counts": label_pair_shared_hash_counts,
        "left_title_quality_counts": left_quality_counts,
        "right_title_quality_counts": right_quality_counts,
        "representative_samples": samples,
    }


def primary_duplicate_conflict_forensics(
    con,
    artifact_id,
    primary_field,
):
    groups = con.execute(
        """SELECT norm_hash, COUNT(*) AS n, COUNT(DISTINCT label) AS label_n
           FROM fingerprints
           WHERE artifact_id=? AND text_field=?
             AND label IS NOT NULL
           GROUP BY norm_hash
           HAVING COUNT(DISTINCT label)>1
           ORDER BY norm_hash""",
        (artifact_id, primary_field),
    ).fetchall()

    group_records = []
    representative_groups = []
    affected_rows = set()
    genuine_groups = 0
    malformed_groups = 0
    normalization_only_groups = 0
    exact_raw_conflict_groups = 0
    label_combination_counts = Counter()

    malformed_categories = {
        "MISSING",
        "EMPTY",
        "KNOWN_PLACEHOLDER",
        "NON_TEXTUAL_GARBAGE",
    }

    for norm_hash, _, _ in groups:
        rows = con.execute(
            """SELECT row_id, raw_hash, label_raw, label,
                      title_quality, raw_text_protected
               FROM fingerprints
               WHERE artifact_id=? AND text_field=? AND norm_hash=?
               ORDER BY row_id, raw_hash""",
            (artifact_id, primary_field, norm_hash),
        ).fetchall()

        for row in rows:
            affected_rows.add(row[0])

        labels = sorted({
            str(r[3])
            for r in rows
            if r[3] is not None
        })
        raw_label_variants = sorted({
            "" if r[2] is None else str(r[2])
            for r in rows
        })
        label_combination_counts["|".join(labels)] += 1

        raw_to_labels = defaultdict(set)
        for row in rows:
            raw_to_labels[row[1]].add(str(row[3]))

        exact_raw_conflict = any(
            len(values) > 1
            for values in raw_to_labels.values()
        )
        normalization_only = (
            not exact_raw_conflict
            and len(raw_to_labels) > 1
        )

        title_quality_counts = Counter(
            (r[4] or "<NONE>")
            for r in rows
        )
        malformed = any(
            category in malformed_categories
            for category in title_quality_counts
        )

        if malformed:
            malformed_groups += 1
        else:
            genuine_groups += 1
        if normalization_only:
            normalization_only_groups += 1
        if exact_raw_conflict:
            exact_raw_conflict_groups += 1

        group_meta = {
            "normalized_sha256": norm_hash,
            "row_count": len(rows),
            "row_ids": sorted({int(r[0]) for r in rows}),
            "labels": labels,
            "raw_label_variants": raw_label_variants,
            "raw_variant_count": len(raw_to_labels),
            "exact_raw_conflict": bool(exact_raw_conflict),
            "normalization_only_conflict": bool(normalization_only),
            "placeholder_or_malformed": bool(malformed),
            "title_quality_counts": {
                key: int(title_quality_counts[key])
                for key in sorted(title_quality_counts)
            },
        }
        group_records.append(group_meta)

        if len(representative_groups) < 10:
            representative_groups.append({
                **group_meta,
                "representative_rows": [
                    {
                        "row_id": r[0],
                        "raw_sha256": r[1],
                        "raw_label": r[2],
                        "canonical_label": r[3],
                        "title_quality": r[4],
                        "excerpt": truncate_diagnostic_text(r[5]),
                    }
                    for r in rows[:6]
                ],
            })

    group_records = sorted(
        group_records,
        key=lambda x: x["normalized_sha256"],
    )
    representative_groups = sorted(
        representative_groups,
        key=lambda x: x["normalized_sha256"],
    )

    return {
        "artifact_id": artifact_id,
        "primary_text_column": primary_field,
        "conflicting_normalized_text_groups": len(groups),
        "affected_rows": len(affected_rows),
        "genuine_nonplaceholder_groups": genuine_groups,
        "placeholder_or_malformed_groups": malformed_groups,
        "exact_raw_conflict_groups": exact_raw_conflict_groups,
        "normalization_only_conflict_groups": normalization_only_groups,
        "label_combination_group_counts": {
            key: int(label_combination_counts[key])
            for key in sorted(label_combination_counts)
        },
        "groups": group_records,
        "representative_groups": representative_groups,
        "full_article_text_persisted": False,
    }


def protected_title_overlap_is_blocking(
    left_field,
    right_field,
    left_primary,
    right_primary,
    normalized_shared_hashes,
):
    return bool(
        left_field == left_primary == "title"
        and right_field == right_primary == "title"
        and int(normalized_shared_hashes) > 0
    )

def protected_near_duplicate_stress(con):
    protected = [
        "banglabait_train",
        "banglabait_dev",
        "banglabait_test",
    ]
    limit = int(
        AUDIT_CONFIG["protected_near_duplicate_sample_per_artifact"]
    )
    threshold = int(
        AUDIT_CONFIG["protected_near_duplicate_hamming_threshold"]
    )

    signatures = {}
    for aid in protected:
        rows = con.execute(
            """SELECT row_id, norm_hash, norm_text_protected
               FROM fingerprints
               WHERE artifact_id=? AND is_primary=1
                 AND norm_text_protected IS NOT NULL
                 AND norm_len>=30
               ORDER BY norm_hash, row_id
               LIMIT ?""",
            (aid, limit),
        ).fetchall()
        signatures[aid] = [
            (row_id, norm_hash, simhash64(text))
            for row_id, norm_hash, text in rows
        ]

    pair_results = {}
    for i, left in enumerate(protected):
        for right in protected[i + 1:]:
            index = [defaultdict(list) for _ in range(4)]
            for row_id, norm_hash, sig in signatures[right]:
                for band in range(4):
                    key = (sig >> (16 * band)) & 0xFFFF
                    index[band][key].append(
                        (row_id, norm_hash, sig)
                    )

            candidate_pairs = set()
            near_pairs = 0
            candidate_cap = 2_000_000
            truncated = False

            for lrow, lhash, lsig in signatures[left]:
                local = set()
                for band in range(4):
                    key = (lsig >> (16 * band)) & 0xFFFF
                    for rrow, rhash, rsig in index[band].get(
                        key,
                        [],
                    ):
                        if lhash == rhash:
                            continue
                        local.add((rrow, rhash, rsig))

                for rrow, rhash, rsig in sorted(
                    local,
                    key=lambda x: (x[0], x[1], x[2]),
                ):
                    pair_key = (lrow, rrow)
                    if pair_key in candidate_pairs:
                        continue
                    candidate_pairs.add(pair_key)
                    if hamming64(lsig, rsig) <= threshold:
                        near_pairs += 1
                    if len(candidate_pairs) >= candidate_cap:
                        truncated = True
                        break
                if truncated:
                    break

            pair_results[(left, right)] = {
                "left_sample_size": len(signatures[left]),
                "right_sample_size": len(signatures[right]),
                "candidate_pairs_checked": len(candidate_pairs),
                "near_pairs_hamming_le_threshold": near_pairs,
                "hamming_threshold": threshold,
                "sample_cap_per_artifact": limit,
                "candidate_cap": candidate_cap,
                "truncated": truncated,
                "gate_blocking": False,
                "interpretation": (
                    "heuristic stress candidates; exact/normalized "
                    "title overlap remains the gate criterion"
                ),
            }
    return pair_results
````

## Full artifact inspection

````python

if globals().get("P0_PREFLIGHT_PASS") is not True:
    raise RuntimeError(
        "Mandatory Kaggle preflight has not passed. "
        "Scientific artifact inspection is blocked."
    )

SCHEMA_ROWS = []
LABEL_ROWS = []
ARTIFACT_SUMMARIES = []
DUPLICATE_ROWS = []
OVERLAP_ROWS = []
NEAR_RESULTS = {}
BLOCKER_FORENSICS = {
    "schema_version": 1,
    "methodology_version": METHODOLOGY_VERSION,
    "protected_field_contract": {
        "dataset": "BanglaBait",
        "primary_text_column": "title",
        "primary_label_column": "label",
    },
    "within_split_primary_conflicts": {},
    "protected_title_quality": {},
    "protected_label_domain": {},
    "protected_split_field_pair_overlaps": {},
}

if INPUT_RECORD:
    db_path = TEMP_ROOT / "p0_fingerprints.sqlite"
    db_path.parent.mkdir(parents=True, exist_ok=True)
    con = initialize_fingerprint_db(db_path)

    for artifact in ARTIFACTS:
        log(f"Inspecting artifact: {artifact['artifact_id']}")
        try:
            schema_rows, label_rows, _, summary = audit_artifact(FROZEN_ROOT, artifact, con)
            SCHEMA_ROWS.extend(schema_rows)
            LABEL_ROWS.extend(label_rows)
            if summary:
                ARTIFACT_SUMMARIES.append(summary)
        except Exception as exc:
            blocking = bool(
                artifact["artifact_id"] in {
                    "banglabait_train", "banglabait_dev", "banglabait_test"
                }
                or artifact["artifact_id"] == "baitbuster_parquet"
            )
            add_finding(
                "BLOCKER" if blocking else "MAJOR",
                "P0-SCHEMA-READ-" + artifact["artifact_id"].upper(),
                "schema",
                f"Artifact inspection failed for {artifact['artifact_id']}.",
                blocking,
                {
                    "artifact": artifact["relative_path"],
                    "error": f"{type(exc).__name__}: {exc}",
                },
            )

    summaries_by_id = {x["artifact_id"]: x for x in ARTIFACT_SUMMARIES}

    # Protected schema/label contract.
    for aid in ["banglabait_train", "banglabait_dev", "banglabait_test"]:
        s = summaries_by_id.get(aid)
        if not s:
            continue
        if s.get("primary_text_column") != "title":
            add_finding(
                "BLOCKER",
                "P0-FIELD-CONTRACT-TEXT-" + aid.upper(),
                "schema",
                f"{aid} did not resolve the required protected primary text column 'title'.",
                True,
                s,
            )
        if s.get("primary_label") != "label":
            add_finding(
                "BLOCKER",
                "P0-FIELD-CONTRACT-LABEL-" + aid.upper(),
                "labels",
                f"{aid} did not resolve the required protected primary label column 'label'.",
                True,
                s,
            )
        if s.get("primary_text_eligible_rows", 0) == 0:
            add_finding(
                "BLOCKER",
                "P0-SCHEMA-TEXT-" + aid.upper(),
                "schema",
                f"{aid} has no usable protected title observations.",
                True,
                s,
            )
        if s.get("primary_label_missing", 0) > 0:
            add_finding(
                "BLOCKER",
                "P0-LABEL-MISSINGROW-" + aid.upper(),
                "labels",
                f"{aid} contains missing protected labels.",
                True,
                {"missing": s["primary_label_missing"], "row_count": s["row_count"]},
            )
        if s.get("primary_label_class_count", 0) < 2:
            add_finding(
                "BLOCKER",
                "P0-LABEL-CLASSCOUNT-" + aid.upper(),
                "labels",
                f"{aid} has fewer than two observed protected label classes.",
                True,
                {"class_count": s.get("primary_label_class_count")},
            )
        if s.get("row_count", 0) <= 0:
            add_finding(
                "BLOCKER",
                "P0-SCHEMA-EMPTY-" + aid.upper(),
                "schema",
                f"{aid} has zero data rows.",
                True,
                s,
            )

    # Independent protected-title quality and raw-label-domain gate.
    for aid in ["banglabait_train", "banglabait_dev", "banglabait_test"]:
        s = summaries_by_id.get(aid)
        if not s:
            continue

        quality = s.get("protected_title_quality")
        label_domain = s.get("protected_label_domain")
        BLOCKER_FORENSICS["protected_title_quality"][aid] = quality
        BLOCKER_FORENSICS["protected_label_domain"][aid] = label_domain

        if quality and int(quality.get("malformed_count", 0)) > 0:
            add_finding(
                "BLOCKER",
                "P0-MALFORMED-PRIMARY-" + aid.upper(),
                "data_quality",
                f"{aid} contains missing/empty/known-placeholder/non-textual protected titles.",
                True,
                quality,
            )

        if label_domain and label_domain.get("pass") is not True:
            add_finding(
                "BLOCKER",
                "P0-LABEL-DOMAIN-" + aid.upper(),
                "labels",
                f"{aid} contains protected labels outside the frozen canonical comparison domain.",
                True,
                label_domain,
            )

    # Generic duplicate summaries remain descriptive for all selected fields.
    DUPLICATE_ROWS = duplicate_report(con, ARTIFACT_SUMMARIES)
    for row in DUPLICATE_ROWS:
        aid = row["artifact_id"]
        primary_field = summaries_by_id.get(aid, {}).get("primary_text_column")
        if row["text_field"] != primary_field:
            continue
        rate = float(row["normalized_duplicate_rate"])
        if rate >= 0.05:
            add_finding(
                "MAJOR",
                "P0-DUP-RATE-" + aid.upper(),
                "duplicates",
                f"{aid} has a normalized protected-primary duplicate rate of at least 5%.",
                False,
                row,
            )
        elif rate > 0:
            add_finding(
                "MINOR",
                "P0-DUP-PRESENT-" + aid.upper(),
                "duplicates",
                f"{aid} contains normalized protected-primary duplicates.",
                False,
                row,
            )

    # Dedicated primary title/label conflict forensics.
    for aid in ["banglabait_train", "banglabait_dev", "banglabait_test"]:
        if aid not in summaries_by_id:
            continue
        forensic = primary_duplicate_conflict_forensics(con, aid, "title")
        BLOCKER_FORENSICS["within_split_primary_conflicts"][aid] = forensic

        if forensic["genuine_nonplaceholder_groups"] > 0:
            add_finding(
                "BLOCKER",
                "P0-DUP-CONFLICT-" + aid.upper(),
                "duplicates",
                f"{aid} contains normalized title groups with conflicting protected labels.",
                True,
                forensic,
            )

    analytical = [
        a for a in ARTIFACTS
        if a["analyze_content"]
        and a["artifact_id"] in summaries_by_id
    ]
    protected_ids = {"banglabait_train", "banglabait_dev", "banglabait_test"}

    for i, left in enumerate(analytical):
        for right in analytical[i + 1:]:
            left_id = left["artifact_id"]
            right_id = right["artifact_id"]
            protected_pair = left_id in protected_ids and right_id in protected_ids

            if protected_pair:
                left_summary = summaries_by_id[left_id]
                right_summary = summaries_by_id[right_id]
                left_primary = left_summary["primary_text_column"]
                right_primary = right_summary["primary_text_column"]

                pair_key = f"{left_id}__{right_id}"
                pair_forensics = {
                    "left_artifact": left_id,
                    "right_artifact": right_id,
                    "left_primary_text_column": left_primary,
                    "right_primary_text_column": right_primary,
                    "left_selected_text_columns": left_summary["selected_text_columns"],
                    "right_selected_text_columns": right_summary["selected_text_columns"],
                    "field_pairs": [],
                }

                secondary_same_field_overlaps = []
                cross_field_overlaps = []

                for left_field in left_summary["selected_text_columns"]:
                    for right_field in right_summary["selected_text_columns"]:
                        rec = field_pair_overlap(
                            con,
                            left_id,
                            right_id,
                            left_field,
                            right_field,
                        )
                        rec["left_primary_text_column"] = left_primary
                        rec["right_primary_text_column"] = right_primary
                        rec["primary_to_primary"] = (
                            left_field == left_primary and right_field == right_primary
                        )
                        rec["stage03_blocking"] = (
                            protected_title_overlap_is_blocking(
                                left_field,
                                right_field,
                                left_primary,
                                right_primary,
                                rec["normalized_shared_hashes"],
                            )
                        )
                        pair_forensics["field_pairs"].append(rec)

                        OVERLAP_ROWS.append({
                            "left_dataset": left["dataset"],
                            "left_artifact": left_id,
                            "left_role": left["role"],
                            "left_text_field": left_field,
                            "left_field_role": rec["left_field_role"],
                            "right_dataset": right["dataset"],
                            "right_artifact": right_id,
                            "right_role": right["role"],
                            "right_text_field": right_field,
                            "right_field_role": rec["right_field_role"],
                            "comparison_scope": (
                                "protected_primary_to_primary"
                                if rec["primary_to_primary"]
                                else (
                                    "protected_secondary_same_field"
                                    if rec["same_field"]
                                    else "protected_cross_field"
                                )
                            ),
                            "exact_raw_shared_hashes": rec["exact_raw_shared_hashes"],
                            "normalized_shared_hashes": rec["normalized_shared_hashes"],
                            "normalization_only_shared_hashes": rec["normalization_only_shared_hashes"],
                            "exact_raw_affected_rows_left": rec["exact_raw_affected_rows_left"],
                            "exact_raw_affected_rows_right": rec["exact_raw_affected_rows_right"],
                            "normalized_affected_rows_left": rec["normalized_affected_rows_left"],
                            "normalized_affected_rows_right": rec["normalized_affected_rows_right"],
                            "normalization_only_affected_rows_left": rec["normalization_only_affected_rows_left"],
                            "normalization_only_affected_rows_right": rec["normalization_only_affected_rows_right"],
                            "left_rows_with_normalized_overlap": rec["normalized_affected_rows_left"],
                            "right_rows_with_normalized_overlap": rec["normalized_affected_rows_right"],
                            "conflicting_label_shared_hashes": rec["conflicting_label_shared_hashes"],
                            "label_pair_shared_hash_counts_json": json.dumps(
                                rec["label_pair_shared_hash_counts"],
                                ensure_ascii=False,
                                sort_keys=True,
                                separators=(",", ":"),
                            ),
                            "left_title_quality_counts_json": json.dumps(
                                rec["left_title_quality_counts"],
                                ensure_ascii=False,
                                sort_keys=True,
                                separators=(",", ":"),
                            ),
                            "right_title_quality_counts_json": json.dumps(
                                rec["right_title_quality_counts"],
                                ensure_ascii=False,
                                sort_keys=True,
                                separators=(",", ":"),
                            ),
                            "stage03_blocking": rec["stage03_blocking"],
                        })

                        if rec["primary_to_primary"] and rec["normalized_shared_hashes"] > 0:
                            add_finding(
                                "BLOCKER",
                                "P0-LEAK-PROTECTED-" + left_id.upper() + "-" + right_id.upper(),
                                "leakage",
                                "Exact or normalized protected-title overlap crosses the BanglaBait train/calibration/test firewall.",
                                True,
                                rec,
                            )
                        elif rec["same_field"] and rec["normalized_shared_hashes"] > 0:
                            secondary_same_field_overlaps.append(rec)
                        elif rec["normalized_shared_hashes"] > 0:
                            cross_field_overlaps.append(rec)

                if secondary_same_field_overlaps:
                    add_finding(
                        "MAJOR",
                        "P0-SECONDARY-SAMEFIELD-" + left_id.upper() + "-" + right_id.upper(),
                        "leakage",
                        "Secondary same-field text overlap exists across protected BanglaBait splits. It is not classified as protected-title leakage under the current title-based task; any future use of the implicated field requires firewall review before training.",
                        False,
                        {
                            "left_artifact": left_id,
                            "right_artifact": right_id,
                            "field_pairs": secondary_same_field_overlaps,
                        },
                    )
                if cross_field_overlaps:
                    add_finding(
                        "INFORMATIONAL",
                        "P0-CROSSFIELD-" + left_id.upper() + "-" + right_id.upper(),
                        "overlap",
                        "Cross-field text overlap exists across protected BanglaBait splits and is retained as diagnostic evidence rather than primary-title leakage.",
                        False,
                        {
                            "left_artifact": left_id,
                            "right_artifact": right_id,
                            "field_pairs": cross_field_overlaps,
                        },
                    )

                BLOCKER_FORENSICS["protected_split_field_pair_overlaps"][pair_key] = pair_forensics

            else:
                # Preserve union-of-selected-fields overlap as descriptive evidence
                # outside the protected primary-title gate.
                exact_shared, norm_shared, left_rows, right_rows, conflict_hashes = artifact_pair_overlap_union(
                    con, left_id, right_id
                )
                train_external_pair = (
                    (left_id == "banglabait_train" and right["dataset"] != "BanglaBait")
                    or (right_id == "banglabait_train" and left["dataset"] != "BanglaBait")
                )
                row = {
                    "left_dataset": left["dataset"],
                    "left_artifact": left_id,
                    "left_role": left["role"],
                    "left_text_field": "*selected_union",
                    "left_field_role": "selected_text_union",
                    "right_dataset": right["dataset"],
                    "right_artifact": right_id,
                    "right_role": right["role"],
                    "right_text_field": "*selected_union",
                    "right_field_role": "selected_text_union",
                    "comparison_scope": (
                        "source_train_vs_external" if train_external_pair else "cross_artifact"
                    ),
                    "exact_raw_shared_hashes": exact_shared,
                    "normalized_shared_hashes": norm_shared,
                    "normalization_only_shared_hashes": None,
                    "exact_raw_affected_rows_left": None,
                    "exact_raw_affected_rows_right": None,
                    "normalized_affected_rows_left": left_rows,
                    "normalized_affected_rows_right": right_rows,
                    "normalization_only_affected_rows_left": None,
                    "normalization_only_affected_rows_right": None,
                    "left_rows_with_normalized_overlap": left_rows,
                    "right_rows_with_normalized_overlap": right_rows,
                    "conflicting_label_shared_hashes": conflict_hashes,
                    "label_pair_shared_hash_counts_json": None,
                    "left_title_quality_counts_json": None,
                    "right_title_quality_counts_json": None,
                    "stage03_blocking": False,
                }
                OVERLAP_ROWS.append(row)
                if train_external_pair and norm_shared > 0:
                    add_finding(
                        "MAJOR",
                        "P0-LEAK-EXTERNAL-" + left_id.upper() + "-" + right_id.upper(),
                        "leakage",
                        "Normalized text overlap exists between the model-development pool and an external diagnostic dataset.",
                        False,
                        row,
                    )
                elif norm_shared > 0:
                    add_finding(
                        "INFORMATIONAL",
                        "P0-OVERLAP-" + left_id.upper() + "-" + right_id.upper(),
                        "overlap",
                        "Normalized text overlap was detected between audited artifacts.",
                        False,
                        row,
                    )

    # Near-duplicate stress now uses the explicit protected primary title field
    # because is_primary is no longer defined heuristically for protected splits.
    NEAR_RESULTS = protected_near_duplicate_stress(con)
    for (left, right), rec in NEAR_RESULTS.items():
        if rec.get("truncated"):
            add_finding(
                "INFORMATIONAL",
                "P0-NEAR-TRUNCATED-" + left.upper() + "-" + right.upper(),
                "leakage",
                "The heuristic protected-title near-duplicate scan reached its deterministic candidate cap.",
                False,
                {"left_artifact": left, "right_artifact": right, **rec},
            )
        if rec["near_pairs_hamming_le_threshold"] > 0:
            add_finding(
                "MAJOR",
                "P0-NEAR-" + left.upper() + "-" + right.upper(),
                "leakage",
                "Heuristic near-duplicate title candidates were detected across protected BanglaBait splits; exact/normalized title overlap remains the automatic gate criterion.",
                False,
                {"left_artifact": left, "right_artifact": right, **rec},
            )

    con.close()
    if db_path.exists():
        db_path.unlink()
    log("Field-attributed schema, label, duplicate, and overlap inspection completed.")
else:
    log("Data inspection skipped because the frozen input did not pass verification.")
````

## Severity classification and P0 decision

Findings are classified as:

- **BLOCKER** — invalidates the Stage 02 gate and forces `P0_NO_GO`;
- **MAJOR** — substantial risk requiring explicit downstream handling but not automatically Stage-03-blocking unless marked as such;
- **MINOR** — limited integrity/reproducibility concern that remains documented;
- **INFORMATIONAL** — descriptive provenance, construct, cross-field, or firewall evidence.

The binary decision is based on `stage03_blocking`.

P0-v1.1.1 does not reduce the strictness of the protected title firewall. It corrects attribution: a protected-title blocker must arise from the explicit protected `title` field rather than from an unqualified union of multiple selected text fields.

````python

END_INPUT_CHECK = {
    "status": "NOT_AVAILABLE",
}

if INPUT_RECORD:
    try:
        after = verify_manifest_root(FROZEN_ROOT)
        if after["content_tree_fingerprint"] != INPUT_RECORD["content_tree_fingerprint"]:
            raise RuntimeError("Analyzed frozen-tree fingerprint changed during audit")

        END_INPUT_CHECK = {
            "status": "PASS",
            "input_mode": INPUT_RECORD["input_mode"],
            "content_tree_fingerprint_before": INPUT_RECORD["content_tree_fingerprint"],
            "content_tree_fingerprint_after": after["content_tree_fingerprint"],
            "raw_input_modified": False,
        }

        if INPUT_RECORD["input_mode"] == "ZIP_MODE":
            current_zip_sha = sha256_file(Path(INPUT_RECORD["zip_path"]))
            if current_zip_sha != INPUT_RECORD["zip_sha256_at_start"]:
                raise RuntimeError(
                    f"Input ZIP changed during audit: {current_zip_sha} != {INPUT_RECORD['zip_sha256_at_start']}"
                )
            END_INPUT_CHECK["zip_sha256_before"] = INPUT_RECORD["zip_sha256_at_start"]
            END_INPUT_CHECK["zip_sha256_after"] = current_zip_sha

        elif INPUT_RECORD["input_mode"] == "EXTRACTED_ROOT_MODE":
            reconstructed_check = (
                verify_extracted_root_against_canonical_zip(
                    FROZEN_ROOT,
                    EXPECTED_ZIP_SHA256,
                    TEMP_ROOT,
                )
            )
            END_INPUT_CHECK["tree_anchor_recheck"] = {
                "status": reconstructed_check["status"],
                "method": reconstructed_check["method"],
                "verified_reconstructed_zip_sha256":
                    reconstructed_check[
                        "verified_reconstructed_zip_sha256"
                    ],
            }

        log("End-of-run read-only input verification: PASS")
    except Exception as exc:
        END_INPUT_CHECK = {
            "status": "FAIL",
            "error": f"{type(exc).__name__}: {exc}",
            "raw_input_modified": "UNKNOWN_OR_CHANGED",
        }
        add_finding(
            "BLOCKER",
            "P0-INTEGRITY-ENDCHECK",
            "integrity",
            "End-of-run frozen-input verification failed.",
            True,
            END_INPUT_CHECK,
        )
        log("End-of-run read-only input verification: FAIL")

# If a new blocking finding was generated by the final input check, update decision outputs.
FINDINGS = sorted(
    FINDINGS,
    key=lambda x: (
        SEVERITY_ORDER[x["severity"]],
        x["code"],
        x["domain"],
        x["message"],
    ),
)
blocking_findings = [x for x in FINDINGS if x["stage03_blocking"]]
P0_DECISION_FINAL = "P0_NO_GO" if blocking_findings else "P0_GO"

if P0_DECISION_FINAL != P0_REPORT["decision"] or FINDINGS != P0_REPORT["findings"]:
    P0_REPORT["decision"] = P0_DECISION_FINAL
    P0_REPORT["findings"] = FINDINGS
    P0_REPORT["stage03_blocking_finding_codes"] = [x["code"] for x in blocking_findings]
    severity_counts = Counter(x["severity"] for x in FINDINGS)
    P0_REPORT["finding_counts"] = {
        severity: severity_counts.get(severity, 0)
        for severity in ["BLOCKER", "MAJOR", "MINOR", "INFORMATIONAL"]
    }
    write_json(RUN_DIR / "P0_AUDIT_REPORT.json", P0_REPORT)

# Rebuild deterministic summary after the final read-only gate.
summary_lines = [
    "# P0 Integrity, Provenance, and Leakage Audit Summary",
    "",
    f"**Decision:** `{P0_DECISION_FINAL}`",
    "",
    f"**Methodology:** `{METHODOLOGY_VERSION}`",
    "",
    f"**Methodology SHA-256:** `{METHODOLOGY_SHA256}`",
    "",
    f"**Frozen package:** `{FREEZE_ID}`",
    "",
    f"**Expected ZIP SHA-256:** `{EXPECTED_ZIP_SHA256}`",
    "",
    "## Finding counts",
    "",
]
for severity in ["BLOCKER", "MAJOR", "MINOR", "INFORMATIONAL"]:
    summary_lines.append(f"- {severity}: {severity_counts.get(severity, 0)}")
summary_lines += ["", "## Stage 03 gate", ""]
if P0_DECISION_FINAL == "P0_GO":
    summary_lines.append(
        "Stage 03 model development is authorized subject to the frozen `SPLIT_FIREWALL.json` and all documented non-blocking findings."
    )
else:
    summary_lines.append(
        "Stage 03 model development remains blocked until every `stage03_blocking=true` finding is resolved through an explicit versioned protocol decision and Stage 02 is re-executed."
    )
summary_lines += ["", "## Blocking findings", ""]
if blocking_findings:
    for item in blocking_findings:
        summary_lines.append(f"- `{item['code']}` — {item['message']}")
else:
    summary_lines.append("- None.")
summary_lines += [
    "",
    "## Construct boundary",
    "",
    "BanglaBait is treated as target-proximal clickbait supervision/proxy rather than direct proof of equivalence with the broader yellow-journalism construct. BanglaClick is external clickbait diagnostic evidence. BanMANI, BanFakeNews-2.0, and BaitBuster-Bangla remain adjacent/boundary constructs and are prohibited from model tuning.",
    "",
    "## BaitBuster serialization rule",
    "",
    "The pinned Version 3 Parquet file is the canonical research artifact. XLSX is secondary reference material. The unavailable publisher-listed CSV is not reconstructed and no cross-serialization byte-equivalence claim is made.",
    "",
]
(RUN_DIR / "P0_AUDIT_SUMMARY.md").write_text(
    "\n".join(summary_lines),
    encoding="utf-8",
    newline="\n",
)

hash_lines = [
    f"Stage: {STAGE_ID}",
    f"Canonical notebook: {CANONICAL_NOTEBOOK}",
    f"Methodology SHA-256: {METHODOLOGY_SHA256}",
    f"Expected Stage 01 ZIP SHA-256: {EXPECTED_ZIP_SHA256}",
    f"Stage 01 notebook SHA-256 (historical provenance identifier): {EXPECTED_STAGE01_NOTEBOOK_SHA256}",
    "Stage 01 notebook provenance status: historical_provenance_identifier_only",
    "Stage 01 notebook runtime physical verification: NOT_PERFORMED",
    "Stage 01 notebook runtime input required: NO",
    f"Tree fingerprint algorithm: {STAGE01_TREE_FINGERPRINT_ALGORITHM}",
    f"Verified Stage 01 tree fingerprint: {INPUT_RECORD['content_tree_fingerprint'] if INPUT_RECORD else 'UNVERIFIED'}",
    f"Tree anchor method: {INPUT_RECORD['tree_anchor']['method'] if INPUT_RECORD else 'UNVERIFIED'}",
    f"Tree anchor canonical ZIP SHA-256: {INPUT_RECORD['tree_anchor'].get('canonical_zip_sha256', INPUT_RECORD['tree_anchor'].get('expected_zip_sha256')) if INPUT_RECORD else 'UNVERIFIED'}",
    f"Split firewall SHA-256: {SPLIT_FIREWALL_SHA256}",
    f"Execution replica ID: {P0_REPLICA_ID}",
    f"Input verification error: {INPUT_VERIFICATION_ERROR or 'NONE'}",
    f"Input mode: {INPUT_RECORD['input_mode'] if INPUT_RECORD else 'UNVERIFIED'}",
    f"End-of-run read-only check: {END_INPUT_CHECK['status']}",
    f"P0 decision: {P0_DECISION_FINAL}",
]
if INPUT_RECORD and INPUT_RECORD.get("zip_sha256"):
    hash_lines.append(f"Verified input ZIP SHA-256: {INPUT_RECORD['zip_sha256']}")
(RUN_DIR / "P0_HASH_VERIFICATION.txt").write_text(
    "\n".join(hash_lines) + "\n",
    encoding="utf-8",
    newline="\n",
)

def package_version(name):
    try:
        return im.version(name)
    except Exception:
        return None

gpu_status = {
    "cuda_visible_devices": os.environ.get("CUDA_VISIBLE_DEVICES"),
    "nvidia_smi_available": False,
    "nvidia_smi_summary": None,
}
try:
    p = subprocess.run(
        ["nvidia-smi", "--query-gpu=name,driver_version", "--format=csv,noheader"],
        capture_output=True,
        text=True,
        timeout=10,
        check=False,
    )
    if p.returncode == 0:
        gpu_status["nvidia_smi_available"] = True
        gpu_status["nvidia_smi_summary"] = p.stdout.strip()
except Exception:
    pass

RUN_METADATA = {
    "schema_version": 1,
    "stage_id": STAGE_ID,
    "canonical_notebook": CANONICAL_NOTEBOOK,
    "execution_id": execution_id,
    "execution_utc": utc_now(),
    "methodology_version": METHODOLOGY_VERSION,
    "methodology_sha256": METHODOLOGY_SHA256,
    "split_firewall_sha256": SPLIT_FIREWALL_SHA256,
    "execution_replica_id": P0_REPLICA_ID,
    "input_mode": INPUT_RECORD["input_mode"] if INPUT_RECORD else None,
    "required_kaggle_dataset_inputs": 1,
    "physical_stage01_notebook_input_required": False,
    "previous_stage02_outputs_used_as_input": False,
    "input_content_tree_fingerprint": (
        INPUT_RECORD["content_tree_fingerprint"] if INPUT_RECORD else None
    ),
    "expected_zip_sha256": EXPECTED_ZIP_SHA256,
    "canonical_environment_spec": {
        "python": EXPECTED_PYTHON_VERSION,
        "packages": EXPECTED_PACKAGE_VERSIONS,
    },
    "environment_verification": (
        INPUT_RECORD["environment_verification"]
        if INPUT_RECORD else None
    ),
    "stage01_provenance": {
        "tree_fingerprint_algorithm": STAGE01_TREE_FINGERPRINT_ALGORITHM,
        "verified_tree_fingerprint": (
            INPUT_RECORD["content_tree_fingerprint"]
            if INPUT_RECORD else None
        ),
        "tree_anchor": (
            INPUT_RECORD["tree_anchor"]
            if INPUT_RECORD else None
        ),
        "stage01_notebook": {
            "sha256": EXPECTED_STAGE01_NOTEBOOK_SHA256,
            "status": "historical_provenance_identifier_only",
            "runtime_physical_verification": False,
            "runtime_input_required": False,
        },
    },
    "python": sys.version,
    "platform": platform.platform(),
    "cpu_count": os.cpu_count(),
    "gpu_status": gpu_status,
    "packages": {
        "pandas": package_version("pandas"),
        "numpy": package_version("numpy"),
        "pyarrow": package_version("pyarrow"),
        "openpyxl": package_version("openpyxl"),
    },
    "determinism": {
        "model_training": False,
        "random_sampling": False,
        "near_duplicate_sample_order": "norm_hash,row_id with sorted capped candidate traversal",
        "output_sorting": "stable deterministic sort",
        "serialized_list_policy": "all persisted lists deterministically ordered",
    },
    "p0_decision": P0_DECISION_FINAL,
    "run_directory": str(RUN_DIR),
}
write_json(RUN_DIR / "P0_RUN_METADATA.json", RUN_METADATA)

(RUN_DIR / "execution.log").write_text(
    "\n".join(EXECUTION_LOG) + "\n",
    encoding="utf-8",
    newline="\n",
)

# Attempt to capture the currently executing Kaggle notebook if the platform
# exposes it as a physical file during execution. Kaggle's saved notebook
# version remains the authoritative executed-notebook artifact.
capture = {
    "status": "KAGGLE_SAVED_VERSION_IS_AUTHORITATIVE",
    "copied_path": None,
    "candidates_checked": [],
}
for candidate in [
    WORK_ROOT / "__notebook__.ipynb",
    WORK_ROOT / CANONICAL_NOTEBOOK,
]:
    capture["candidates_checked"].append(str(candidate))
    if candidate.is_file():
        target = RUN_DIR / CANONICAL_NOTEBOOK
        shutil.copyfile(candidate, target)
        capture["status"] = "PHYSICAL_NOTEBOOK_COPY_CAPTURED"
        capture["copied_path"] = str(target)
        break
write_json(RUN_DIR / "EXECUTED_NOTEBOOK_CAPTURE_STATUS.json", capture)

# Deterministic scientific outputs are hashed separately for Track A/B
# reproducibility comparison. Run metadata, timestamps, and logs are excluded.
deterministic_names = [
    "P0_AUDIT_REPORT.json",
    "P0_AUDIT_SUMMARY.md",
    "P0_BLOCKER_FORENSICS.json",
    "PROTECTED_TITLE_QUALITY_REPORT.csv",
    "DATASET_SCHEMA_REPORT.csv",
    "LABEL_DISTRIBUTION_REPORT.csv",
    "DUPLICATE_REPORT.csv",
    "CROSS_DATASET_OVERLAP_REPORT.csv",
    "PROVENANCE_DEPENDENCY_GRAPH.json",
    "CONSTRUCT_ALIGNMENT_MATRIX.csv",
    "SPLIT_FIREWALL.json",
]
deterministic_hashes = {
    name: sha256_file(RUN_DIR / name)
    for name in deterministic_names
    if (RUN_DIR / name).is_file()
}
(RUN_DIR / "P0_DETERMINISTIC_OUTPUT_SHA256SUMS.txt").write_text(
    "\n".join(
        f"{digest}  {name}"
        for name, digest in sorted(deterministic_hashes.items())
    ) + "\n",
    encoding="utf-8",
    newline="\n",
)

# Hash all output artifacts after notebook-capture status and deterministic
# checksum controls have been written.
output_hashes = {}
for p in sorted(RUN_DIR.iterdir()):
    if p.is_file() and p.name != "P0_OUTPUT_SHA256SUMS.txt":
        output_hashes[p.name] = sha256_file(p)

(RUN_DIR / "P0_OUTPUT_SHA256SUMS.txt").write_text(
    "\n".join(
        f"{digest}  {name}"
        for name, digest in sorted(output_hashes.items())
    ) + "\n",
    encoding="utf-8",
    newline="\n",
)

print("=" * 80)
print("FINAL STAGE 02 DECISION:", P0_DECISION_FINAL)
print("Execution ID:", execution_id)
print("Results:", RUN_DIR)
print("Methodology SHA-256:", METHODOLOGY_SHA256)
print("=" * 80)
````
